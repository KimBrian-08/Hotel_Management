#### 6. OOP Concepts Applied


- Classes and Objects -> Classes such as Guest, Room, Reservation, and Hotel are used to create objects that represent real hotel entities.

- Encapsulation -> Sensitive or internal data is protected using private attributes such as `_status`, `_services`, `_guests`, and `_rooms`. Properties control access.

- Inheritance -> Guest and Staff inherit from Person. StandardRoom, DeluxeRoom, and Suite inherit from Room.

- Method Overriding -> Child classes override methods such as `get_role()`, `get_description()`, and `calculate_cost()`.

- Polymorphism -> Different room types calculate cost differently using the same method name: `calculate_cost(nights)`.

- Abstraction -> Person and Room are abstract base classes. They define abstract methods that child classes must implement.

- Composition -> A Reservation contains a Guest object, a Room object, and HotelService objects. The Hotel class contains collections of guests, rooms, reservations, and services.

- Class Methods -> `Guest.number_of_guests()` is a class method.

- Static Methods -> `Guest.validate_phone()` and `Room.validate_nights()` are static methods.

- Special Methods -> Special methods such as `__str__()`, `__repr__()`, `__len__()`, `__eq__()`, and `__hash__()` are implemented.

- Exception Handling -> Custom exceptions such as `RoomUnavailableError`, `GuestNotFoundError`, `InvalidNightsError`, and `InvalidReservationStatusError` are used.

````python
"""
Custom exceptions for the Hotel Reservation and Management System.

This module demonstrates exception handling and custom exception design.
"""


class HotelError(Exception):
    """Base exception for all hotel-related errors."""
    pass


class ValidationError(HotelError):
    """Raised when input data is invalid."""
    pass


class GuestNotFoundError(HotelError):
    """Raised when a guest cannot be found."""
    pass


class DuplicateGuestError(HotelError):
    """Raised when a guest already exists in the system."""
    pass


class RoomNotFoundError(HotelError):
    """Raised when a room cannot be found."""
    pass


class RoomUnavailableError(HotelError):
    """Raised when a room cannot be reserved."""
    pass


class InvalidNightsError(HotelError):
    """Raised when the number of nights is invalid."""
    pass


class DuplicateReservationError(HotelError):
    """Raised when a duplicate reservation is attempted."""
    pass


class ReservationNotFoundError(HotelError):
    """Raised when a reservation cannot be found."""
    pass


class ServiceNotFoundError(HotelError):
    """Raised when a hotel service cannot be found."""
    pass


class InvalidReservationStatusError(HotelError):
    """Raised when a reservation operation is not allowed due to its status."""
    pass


class PaymentError(HotelError):
    """Raised when a payment operation fails."""
    pass
````

##  PERSON, GUEST, & STAFF

````python
"""
Person, Guest, and Staff classes.

This module demonstrates:
- Abstraction
- Inheritance
- Encapsulation
- Class attributes
- Class methods
- Static methods
- Special methods
"""


class Person(ABC):
    """
    Abstract base class for all people in the hotel system.

    Demonstrates abstraction through the abstract method get_role().
    """

    def __init__(self, name, id_number):
        self.name = name
        self.id_number = id_number

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Name must be a non-empty string.")
        self._name = value.strip()

    @property
    def id_number(self):
        return self._id_number

    @id_number.setter
    def id_number(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("ID number must be a non-empty string.")
        self._id_number = value.strip()

    @abstractmethod
    def get_role(self):
        """Return the role of the person in the hotel system."""
        pass

    def get_details(self):
        """Return basic person details."""
        return f"{self.get_role()}: {self.name} (ID: {self.id_number})"

    def __str__(self):
        return self.get_details()

    def __repr__(self):
        return f"<{self.__class__.__name__}(name='{self.name}', id_number='{self.id_number}')>"


class Guest(Person):
    """
    Represents a hotel guest.

    Inherits from Person.
    """

    guest_count = 0

    def __init__(self, name, id_number, phone, guest_type="Regular"):
        super().__init__(name, id_number)
        self.phone = Guest.validate_phone(phone)
        self.guest_type = guest_type
        self.registered_on = datetime.now().date()
        self._loyalty_points = 0
        Guest.guest_count += 1

    @property
    def phone(self):
        return self._phone

    @phone.setter
    def phone(self, value):
        self._phone = Guest.validate_phone(value)

    @property
    def guest_type(self):
        return self._guest_type

    @guest_type.setter
    def guest_type(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Guest type must be a non-empty string.")
        self._guest_type = value.strip()

    @property
    def loyalty_points(self):
        return self._loyalty_points

    def add_loyalty_points(self, points):
        """Add loyalty points to the guest account."""
        if not isinstance(points, int) or points < 0:
            raise ValueError("Loyalty points must be zero or a positive integer.")
        self._loyalty_points += points

    def get_role(self):
        return "Guest"

    @classmethod
    def number_of_guests(cls):
        """Return the total number of registered guests."""
        return cls.guest_count

    @staticmethod
    def validate_phone(phone):
        """Validate phone number format."""
        if not isinstance(phone, str):
            raise ValueError("Phone number must be a string.")

        phone = phone.strip()

        if len(phone) < 7:
            raise ValueError("Phone number is too short.")

        return phone

    def __str__(self):
        return (
            f"Guest: {self.name} | ID: {self.id_number} | "
            f"Phone: {self.phone} | Type: {self.guest_type}"
        )


class Staff(Person):
    """
    Represents hotel staff.

    Inherits from Person.
    """

    def __init__(self, name, id_number, role):
        super().__init__(name, id_number)
        self.role = role

    @property
    def role(self):
        return self._role

    @role.setter
    def role(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Staff role must be a non-empty string.")
        self._role = value.strip()

    def get_role(self):
        return f"Staff - {self.role}"

    def __str__(self):
        return f"Staff: {self.name} | ID: {self.id_number} | Role: {self.role}"
````

## ROOMS

````python
"""
Room classes.

This module demonstrates:
- Abstraction
- Inheritance
- Polymorphism
- Method overriding
- Encapsulation
- Static methods
- Special methods
"""


class Room(ABC):
    """
    Abstract base class for all hotel rooms.

    Child classes must implement:
    - get_description()
    - calculate_cost(nights)
    """

    AVAILABLE = "AVAILABLE"
    RESERVED = "RESERVED"
    OCCUPIED = "OCCUPIED"
    CLEANING = "CLEANING"
    MAINTENANCE = "MAINTENANCE"

    def __init__(self, room_number, base_price):
        self.room_number = room_number
        self.base_price = base_price
        self._status = Room.AVAILABLE

    @property
    def room_number(self):
        return self._room_number

    @room_number.setter
    def room_number(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Room number must be a non-empty string.")
        self._room_number = value.strip()

    @property
    def base_price(self):
        return self._base_price

    @base_price.setter
    def base_price(self, value):
        if not isinstance(value, (int, float)) or value <= 0:
            raise ValueError("Base price must be greater than zero.")
        self._base_price = float(value)

    @property
    def status(self):
        return self._status

    def is_available(self):
        """Check whether the room is available for booking."""
        return self._status == Room.AVAILABLE

    def make_reserved(self):
        """Reserve the room."""
        if not self.is_available():
            raise RoomUnavailableError(
                f"Room {self.room_number} is not available for reservation."
            )
        self._status = Room.RESERVED

    def make_occupied(self):
        """Mark the room as occupied."""
        if self._status != Room.RESERVED:
            raise RoomUnavailableError(
                f"Room {self.room_number} must be reserved before check-in."
            )
        self._status = Room.OCCUPIED

    def make_cleaning(self):
        """Mark the room as cleaning."""
        self._status = Room.CLEANING

    def make_available(self):
        """Mark the room as available."""
        self._status = Room.AVAILABLE

    @staticmethod
    def validate_nights(nights):
        """Validate number of nights."""
        if isinstance(nights, bool) or not isinstance(nights, int):
            raise InvalidNightsError("Number of nights must be an integer.")

        if nights <= 0 or nights > 365:
            raise InvalidNightsError(
                "Number of nights must be a positive integer not greater than 365."
            )

        return nights

    @abstractmethod
    def get_description(self):
        """Return a description of the room."""
        pass

    @abstractmethod
    def calculate_cost(self, nights):
        """Calculate the accommodation cost for the room."""
        pass

    def __str__(self):
        return (
            f"Room {self.room_number} | Status: {self.status} | "
            f"Base Price: {self.base_price:.2f}"
        )

    def __repr__(self):
        return (
            f"<{self.__class__.__name__}(room_number='{self.room_number}', "
            f"status='{self.status}')>"
        )


class StandardRoom(Room):
    """
    Represents a standard hotel room.
    """

    def __init__(self, room_number, base_price, has_tv=True):
        super().__init__(room_number, base_price)
        self.has_tv = has_tv

    def get_description(self):
        tv_info = "with TV" if self.has_tv else "without TV"
        return f"Standard Room {self.room_number} {tv_info}"

    def calculate_cost(self, nights):
        Room.validate_nights(nights)
        return round(self.base_price * nights, 2)


class DeluxeRoom(Room):
    """
    Represents a deluxe hotel room.
    """

    def __init__(self, room_number, base_price, includes_breakfast=True):
        super().__init__(room_number, base_price)
        self.includes_breakfast = includes_breakfast

    def get_description(self):
        breakfast_info = "with breakfast" if self.includes_breakfast else "without breakfast"
        return f"Deluxe Room {self.room_number} {breakfast_info}"

    def calculate_cost(self, nights):
        Room.validate_nights(nights)
        return round(self.base_price * nights * 1.15, 2)


class Suite(Room):
    """
    Represents a luxury hotel suite.
    """

    def __init__(self, room_number, base_price, lounge_access=True):
        super().__init__(room_number, base_price)
        self.lounge_access = lounge_access

    def get_description(self):
        lounge_info = "with lounge access" if self.lounge_access else "without lounge access"
        return f"Executive Suite {self.room_number} {lounge_info}"

    def calculate_cost(self, nights):
        Room.validate_nights(nights)

        cost = self.base_price * nights * 1.25

        if self.lounge_access:
            cost += 500 * nights

        return round(cost, 2)
````

## HOTEL SERVICES

````python
"""
HotelService class.

This module demonstrates:
- Encapsulation
- Class attributes
- Class methods
- Special methods
"""


class HotelService:
    """
    Represents an additional hotel service such as laundry,
    breakfast, airport pickup, or room service.
    """

    _next_service_id = 100

    def __init__(self, name, price):
        self.service_id = HotelService._generate_service_id()
        self.name = name
        self.price = price

    @classmethod
    def _generate_service_id(cls):
        cls._next_service_id += 1
        return f"SRV-{cls._next_service_id}"

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Service name must be a non-empty string.")
        self._name = value.strip()

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if not isinstance(value, (int, float)) or value < 0:
            raise ValueError("Service price must be zero or greater.")
        self._price = float(value)

    def __str__(self):
        return f"{self.service_id}: {self.name} - {self.price:.2f}"

    def __repr__(self):
        return f"<HotelService(service_id='{self.service_id}', name='{self.name}')>"

    def __eq__(self, other):
        if not isinstance(other, HotelService):
            return NotImplemented
        return self.service_id == other.service_id

    def __hash__(self):
        return hash(self.service_id)
````

## PAYMENT

````python
"""
Payment classes.

This module demonstrates:
- Abstraction
- Inheritance
- Polymorphism
- Encapsulation
- Class attributes
- Class methods
- Static methods
- Special methods
"""


class Payment(ABC):
    """
    Abstract base class for all payment types.

    Child classes must implement:
    - get_payment_method()
    - process_payment()
    """

    _next_payment_id = 5000

    PENDING = "PENDING"
    COMPLETED = "COMPLETED"
    FAILED = "FAILED"
    REFUNDED = "REFUNDED"

    def __init__(self, amount):
        self._payment_id = Payment._generate_payment_id()
        self.amount = amount
        self._payment_date = datetime.now()
        self._status = Payment.PENDING

    @classmethod
    def _generate_payment_id(cls):
        cls._next_payment_id += 1
        return f"PAY-{cls._next_payment_id}"

    @property
    def payment_id(self):
        return self._payment_id

    @property
    def payment_date(self):
        return self._payment_date

    @property
    def status(self):
        return self._status

    @property
    def amount(self):
        return self._amount

    @amount.setter
    def amount(self, value):
        if isinstance(value, bool) or not isinstance(value, (int, float)):
            raise PaymentError("Payment amount must be a number.")

        if value <= 0:
            raise PaymentError("Payment amount must be greater than zero.")

        self._amount = float(value)

    @abstractmethod
    def get_payment_method(self):
        """Return the payment method name."""
        pass

    @abstractmethod
    def process_payment(self):
        """Process the payment."""
        pass

    def _base_receipt_lines(self):
        return [
            "=" * 45,
            "PAYMENT RECEIPT",
            "=" * 45,
            f"Payment ID: {self.payment_id}",
            f"Payment Method: {self.get_payment_method()}",
            f"Amount Paid: {self.amount:.2f}",
            f"Payment Status: {self.status}",
            f"Payment Date: {self.payment_date.strftime('%Y-%m-%d %H:%M:%S')}"
        ]

    def _additional_receipt_lines(self):
        """Child classes can override this method to add extra receipt details."""
        return []

    def get_receipt(self):
        """Return a formatted payment receipt."""
        lines = self._base_receipt_lines()
        lines.extend(self._additional_receipt_lines())
        lines.append("=" * 45)
        return "\n".join(lines)

    def __str__(self):
        return (
            f"{self.payment_id} | {self.get_payment_method()} | "
            f"Amount: {self.amount:.2f} | Status: {self.status}"
        )

    def __repr__(self):
        return (
            f"<{self.__class__.__name__}(payment_id='{self.payment_id}', "
            f"amount={self.amount}, status='{self.status}')>"
        )


class CashPayment(Payment):
    """
    Represents cash payment.
    """

    def __init__(self, amount, cash_received=None):
        super().__init__(amount)
        self.cash_received = cash_received if cash_received is not None else amount

    @property
    def cash_received(self):
        return self._cash_received

    @cash_received.setter
    def cash_received(self, value):
        if isinstance(value, bool) or not isinstance(value, (int, float)):
            raise PaymentError("Cash received must be a number.")

        if value < 0:
            raise PaymentError("Cash received cannot be negative.")

        self._cash_received = float(value)

    @property
    def change_due(self):
        if self.cash_received >= self.amount:
            return round(self.cash_received - self.amount, 2)
        return 0.00

    def get_payment_method(self):
        return "Cash"

    def process_payment(self):
        if self.cash_received < self.amount:
            self._status = Payment.FAILED
            raise PaymentError(
                "Insufficient cash received to complete this payment."
            )

        self._status = Payment.COMPLETED
        return self.get_receipt()

    def _additional_receipt_lines(self):
        return [
            f"Cash Received: {self.cash_received:.2f}",
            f"Change Due: {self.change_due:.2f}"
        ]


class CardPayment(Payment):
    """
    Represents card payment.

    This is a simulation only. In a real system, card details
    should be tokenized using a payment provider such as Stripe.
    """

    def __init__(self, amount, card_number, expiry_date, cvv):
        super().__init__(amount)
        self._card_number = CardPayment.validate_card_number(card_number)
        self._expiry_date = CardPayment.validate_expiry_date(expiry_date)
        self._cvv = CardPayment.validate_cvv(cvv)

    @property
    def masked_card_number(self):
        last_four_digits = self._card_number[-4:]
        return f"****-****-****-{last_four_digits}"

    @staticmethod
    def validate_card_number(card_number):
        """Validate card number format."""
        if not isinstance(card_number, str):
            raise PaymentError("Card number must be a string.")

        cleaned_card_number = card_number.replace(" ", "").replace("-", "")

        if not cleaned_card_number.isdigit():
            raise PaymentError("Card number must contain only digits.")

        if len(cleaned_card_number) < 12 or len(cleaned_card_number) > 19:
            raise PaymentError("Card number must be between 12 and 19 digits.")

        return cleaned_card_number

    @staticmethod
    def validate_expiry_date(expiry_date):
        """Validate card expiry date in MM/YY format."""
        if not isinstance(expiry_date, str):
            raise PaymentError("Expiry date must be a string.")

        expiry_date = expiry_date.strip()
        parts = expiry_date.split("/")

        if len(parts) != 2:
            raise PaymentError("Expiry date must be in MM/YY format.")

        month_string = parts[0].strip()
        year_string = parts[1].strip()

        if len(month_string) != 2 or len(year_string) != 2:
            raise PaymentError("Expiry date must be in MM/YY format.")

        try:
            month = int(month_string)
            year = int(year_string)
        except ValueError:
            raise PaymentError("Expiry date must contain valid numbers.")

        if month < 1 or month > 12:
            raise PaymentError("Expiry month must be between 01 and 12.")

        now = datetime.now()
        current_year = now.year % 100
        current_month = now.month

        if year < current_year or (year == current_year and month < current_month):
            raise PaymentError("Card is expired.")

        return expiry_date

    @staticmethod
    def validate_cvv(cvv):
        """Validate card CVV."""
        if not isinstance(cvv, str):
            raise PaymentError("CVV must be a string.")

        cvv = cvv.strip()

        if not cvv.isdigit():
            raise PaymentError("CVV must contain only digits.")

        if len(cvv) < 3 or len(cvv) > 4:
            raise PaymentError("CVV must be 3 or 4 digits.")

        return cvv

    def get_payment_method(self):
        return "Card"

    def process_payment(self):
        """
        Simulate card authorization.

        In a production system, this would call a payment gateway
        such as Stripe, PayPal, or a bank API.
        """
        if not self._card_number or not self._cvv:
            self._status = Payment.FAILED
            raise PaymentError("Card authorization failed.")

        self._status = Payment.COMPLETED
        return self.get_receipt()

    def _additional_receipt_lines(self):
        return [
            f"Card Number: {self.masked_card_number}",
            f"Expiry Date: {self._expiry_date}"
        ]


class MobilePayment(Payment):
    """
    Represents mobile money payment such as M-Pesa, Airtel Money,
    MTN Mobile Money, or other mobile wallet providers.
    """

    def __init__(self, amount, phone_number, provider):
        super().__init__(amount)
        self.phone_number = phone_number
        self.provider = provider

    @property
    def phone_number(self):
        return self._phone_number

    @phone_number.setter
    def phone_number(self, value):
        if not isinstance(value, str):
            raise PaymentError("Phone number must be a string.")

        value = value.strip()

        if len(value) < 10:
            raise PaymentError("Phone number is too short.")

        self._phone_number = value

    @property
    def provider(self):
        return self._provider

    @provider.setter
    def provider(self, value):
        if not isinstance(value, str) or not value.strip():
            raise PaymentError("Mobile payment provider must be a non-empty string.")

        self._provider = value.strip()

    def get_payment_method(self):
        return f"Mobile - {self.provider}"

    def process_payment(self):
        """
        Simulate mobile money STK push or wallet authorization.
        """
        if len(self.phone_number) < 10 or not self.provider:
            self._status = Payment.FAILED
            raise PaymentError("Mobile payment authorization failed.")

        self._status = Payment.COMPLETED
        return self.get_receipt()

    def _additional_receipt_lines(self):
        return [
            f"Provider: {self.provider}",
            f"Phone Number: {self.phone_number}"
        ]
````

## RESERVATION

````python
"""
Reservation class.

This module demonstrates:
- Composition
- Encapsulation
- Exception handling
- Special methods
- Class methods
"""


class Reservation:
    """
    Represents a reservation made by a guest for a room.

    Demonstrates composition because a Reservation contains:
    - Guest object
    - Room object
    - HotelService objects
    - Payment objects
    """

    _next_reservation_id = 1000

    PENDING = "PENDING"
    ACTIVE = "ACTIVE"
    CANCELLED = "CANCELLED"
    COMPLETED = "COMPLETED"

    UNPAID = "UNPAID"
    PARTIALLY_PAID = "PARTIALLY_PAID"
    PAID = "PAID"

    def __init__(self, guest, room, nights, check_in_date=None):
        if not isinstance(guest, Guest):
            raise TypeError("guest must be an instance of Guest.")

        if not isinstance(room, Room):
            raise TypeError("room must be an instance of Room.")

        Room.validate_nights(nights)

        if not room.is_available():
            raise RoomUnavailableError(
                f"Room {room.room_number} is not available for reservation."
            )

        self._reservation_id = Reservation._generate_reservation_id()
        self._guest = guest
        self._room = room
        self.nights = nights
        self._check_in_date = check_in_date if check_in_date else datetime.now().date()
        self._status = Reservation.PENDING
        self._services = []
        self._payments = []
        self._payment_status = Reservation.UNPAID
        self._created_on = datetime.now()

        self._room.make_reserved()

    @classmethod
    def _generate_reservation_id(cls):
        cls._next_reservation_id += 1
        return f"RES-{cls._next_reservation_id}"

    @property
    def reservation_id(self):
        return self._reservation_id

    @property
    def guest(self):
        return self._guest

    @property
    def room(self):
        return self._room

    @property
    def check_in_date(self):
        return self._check_in_date

    @property
    def status(self):
        return self._status

    @property
    def created_on(self):
        return self._created_on

    @property
    def services(self):
        """Return a tuple of services to prevent external modification."""
        return tuple(self._services)

    @property
    def payments(self):
        """Return a tuple of payments to prevent external modification."""
        return tuple(self._payments)

    @property
    def payment_status(self):
        return self._payment_status

    @property
    def nights(self):
        return self._nights

    @nights.setter
    def nights(self, value):
        Room.validate_nights(value)
        self._nights = value

    @property
    def amount_paid(self):
        total_paid = sum(
            payment.amount for payment in self._payments
            if payment.status == Payment.COMPLETED
        )
        return round(total_paid, 2)

    @property
    def balance_due(self):
        balance = self.calculate_total_charge() - self.amount_paid
        return round(max(balance, 0), 2)

    @property
    def is_paid(self):
        return self.balance_due <= 0

    def add_service(self, service):
        """Add an additional hotel service to the reservation."""
        if not isinstance(service, HotelService):
            raise TypeError("service must be an instance of HotelService.")

        if self._status in [Reservation.CANCELLED, Reservation.COMPLETED]:
            raise InvalidReservationStatusError(
                "Services cannot be added to a cancelled or completed reservation."
            )

        if service in self._services:
            raise ValueError(
                f"Service {service.name} has already been added to this reservation."
            )

        self._services.append(service)

    def record_payment(self, payment):
        """Record a completed payment against this reservation."""
        if not isinstance(payment, Payment):
            raise TypeError("payment must be an instance of Payment.")

        if self._status == Reservation.CANCELLED:
            raise InvalidReservationStatusError(
                "Payments cannot be recorded for a cancelled reservation."
            )

        if payment.status != Payment.COMPLETED:
            raise PaymentError("Only completed payments can be recorded.")

        projected_total_paid = round(self.amount_paid + payment.amount, 2)
        total_charge = round(self.calculate_total_charge(), 2)

        if projected_total_paid > total_charge + 0.01:
            raise PaymentError("Payment amount exceeds outstanding balance.")

        self._payments.append(payment)
        self._update_payment_status()

    def _update_payment_status(self):
        """Update payment status based on amount paid."""
        total_charge = self.calculate_total_charge()

        if self.amount_paid <= 0:
            self._payment_status = Reservation.UNPAID
        elif self.amount_paid >= total_charge:
            self._payment_status = Reservation.PAID
        else:
            self._payment_status = Reservation.PARTIALLY_PAID

    def calculate_room_charge(self):
        """Calculate accommodation charge only."""
        return self._room.calculate_cost(self.nights)

    def calculate_service_charge(self):
        """Calculate total service charge."""
        total_service_charge = sum(service.price for service in self._services)
        return round(total_service_charge, 2)

    def calculate_total_charge(self):
        """Calculate total bill amount."""
        total = self.calculate_room_charge() + self.calculate_service_charge()
        return round(total, 2)

    def cancel(self):
        """Cancel the reservation."""
        if self._status == Reservation.CANCELLED:
            raise InvalidReservationStatusError(
                f"Reservation {self.reservation_id} is already cancelled."
            )

        if self._status == Reservation.COMPLETED:
            raise InvalidReservationStatusError(
                f"Reservation {self.reservation_id} is completed and cannot be cancelled."
            )

        self._status = Reservation.CANCELLED
        self._room.make_available()

    def check_in(self):
        """Check the guest into the reservation."""
        if self._status == Reservation.CANCELLED:
            raise InvalidReservationStatusError(
                f"Reservation {self.reservation_id} is cancelled and cannot check in."
            )

        if self._status == Reservation.COMPLETED:
            raise InvalidReservationStatusError(
                f"Reservation {self.reservation_id} is already completed."
            )

        self._room.make_occupied()
        self._status = Reservation.ACTIVE

    def check_out(self):
        """Check the guest out of the reservation."""
        if self._status != Reservation.ACTIVE:
            raise InvalidReservationStatusError(
                f"Reservation {self.reservation_id} must be active before check-out."
            )

        self._room.make_available()
        self._status = Reservation.COMPLETED

    def get_bill_summary(self):
        """Generate a formatted final bill summary."""
        lines = []

        lines.append("=" * 45)
        lines.append(f"FINAL BILL - {self.reservation_id}")
        lines.append("=" * 45)
        lines.append(f"Guest: {self.guest.name}")
        lines.append(f"Guest ID: {self.guest.id_number}")
        lines.append(f"Room: {self.room.room_number}")
        lines.append(f"Room Type: {self.room.get_description()}")
        lines.append(f"Nights: {self.nights}")
        lines.append(f"Room Charge: {self.calculate_room_charge():.2f}")

        if self._services:
            lines.append("Services:")
            for service in self._services:
                lines.append(f"  - {service.name}: {service.price:.2f}")
        else:
            lines.append("Services: None")

        lines.append(f"Service Charge: {self.calculate_service_charge():.2f}")
        lines.append("-" * 45)
        lines.append(f"Total Charge: {self.calculate_total_charge():.2f}")
        lines.append(f"Payment Status: {self.payment_status}")
        lines.append(f"Amount Paid: {self.amount_paid:.2f}")
        lines.append(f"Balance Due: {self.balance_due:.2f}")
        lines.append("=" * 45)

        return "\n".join(lines)

    def __len__(self):
        """Return the number of nights in the reservation."""
        return self.nights

    def __str__(self):
        return (
            f"{self.reservation_id} | {self.guest.name} | "
            f"Room {self.room.room_number} | {self.nights} nights | "
            f"{self.status} | Payment: {self.payment_status} | "
            f"Total: {self.calculate_total_charge():.2f}"
        )

    def __repr__(self):
        return (
            f"<Reservation(reservation_id='{self.reservation_id}', "
            f"guest='{self.guest.name}', room='{self.room.room_number}')>"
        )
````

## HOTEL CONTROLLER

````python
"""
Hotel controller class.

This module demonstrates:
- Composition
- Encapsulation
- Exception handling
- Controller class design
"""


class Hotel:
    """
    Central controller class for the hotel system.

    Manages:
    - Guests
    - Staff
    - Rooms
    - Reservations
    - Hotel services
    - Payments
    """

    def __init__(self, name):
        self.name = name
        self._guests = []
        self._staff = []
        self._rooms = []
        self._reservations = []
        self._services = []

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Hotel name must be a non-empty string.")
        self._name = value.strip()

    @property
    def guests(self):
        return tuple(self._guests)

    @property
    def staff(self):
        return tuple(self._staff)

    @property
    def rooms(self):
        return tuple(self._rooms)

    @property
    def reservations(self):
        return tuple(self._reservations)

    @property
    def services(self):
        return tuple(self._services)

    def register_guest(self, name, id_number, phone, guest_type="Regular"):
        """Register a new guest."""
        for guest in self._guests:
            if guest.id_number == id_number:
                raise DuplicateGuestError(
                    f"Guest with ID {id_number} already exists."
                )

        guest = Guest(name, id_number, phone, guest_type)
        self._guests.append(guest)
        return guest

    def register_staff(self, name, id_number, role):
        """Register a new staff member."""
        for staff in self._staff:
            if staff.id_number == id_number:
                raise ValueError(f"Staff with ID {id_number} already exists.")

        staff = Staff(name, id_number, role)
        self._staff.append(staff)
        return staff

    def add_room(self, room_type, room_number, base_price):
        """Add a room to the hotel."""
        for room in self._rooms:
            if room.room_number == room_number:
                raise ValueError(f"Room {room_number} already exists.")

        if not isinstance(room_type, str) or not room_type.strip():
            raise ValueError("Room type must be a non-empty string.")

        room_type = room_type.strip().lower()

        if room_type == "standard":
            room = StandardRoom(room_number, base_price)
        elif room_type == "deluxe":
            room = DeluxeRoom(room_number, base_price)
        elif room_type == "suite":
            room = Suite(room_number, base_price)
        else:
            raise ValueError(
                "Invalid room type. Use 'standard', 'deluxe', or 'suite'."
            )

        self._rooms.append(room)
        return room

    def add_service(self, name, price):
        """Add a hotel service."""
        service = HotelService(name, price)
        self._services.append(service)
        return service

    def get_guest_by_id(self, id_number):
        """Retrieve a guest by ID."""
        id_number = str(id_number).strip()

        for guest in self._guests:
            if guest.id_number == id_number:
                return guest

        raise GuestNotFoundError(f"Guest with ID {id_number} was not found.")

    def get_room_by_number(self, room_number):
        """Retrieve a room by room number."""
        room_number = str(room_number).strip()

        for room in self._rooms:
            if room.room_number == room_number:
                return room

        raise RoomNotFoundError(f"Room {room_number} was not found.")

    def get_reservation_by_id(self, reservation_id):
        """Retrieve a reservation by reservation ID."""
        reservation_id = str(reservation_id).strip()

        for reservation in self._reservations:
            if reservation.reservation_id == reservation_id:
                return reservation

        raise ReservationNotFoundError(
            f"Reservation {reservation_id} was not found."
        )

    def get_service_by_id(self, service_id):
        """Retrieve a service by service ID."""
        service_id = str(service_id).strip()

        for service in self._services:
            if service.service_id == service_id:
                return service

        raise ServiceNotFoundError(f"Service {service_id} was not found.")

    def search_rooms(self, room_number=None, room_type=None, available_only=True):
        """
        Search rooms by room number, room type, and availability.
        """
        results = self._rooms.copy()

        if room_number:
            room_number = str(room_number).strip()
            results = [
                room for room in results
                if room.room_number == room_number
            ]

        if room_type:
            if not isinstance(room_type, str) or not room_type.strip():
                raise ValueError("Room type must be a non-empty string.")

            room_type = room_type.strip().lower()

            if room_type == "standard":
                results = [
                    room for room in results
                    if isinstance(room, StandardRoom)
                ]
            elif room_type == "deluxe":
                results = [
                    room for room in results
                    if isinstance(room, DeluxeRoom)
                ]
            elif room_type == "suite":
                results = [
                    room for room in results
                    if isinstance(room, Suite)
                ]
            else:
                raise ValueError(
                    "Invalid room type. Use 'standard', 'deluxe', or 'suite'."
                )

        if available_only:
            results = [room for room in results if room.is_available()]

        return results

    def make_reservation(self, guest_id, room_number, nights):
        """Create a reservation."""
        guest = self.get_guest_by_id(guest_id)
        room = self.get_room_by_number(room_number)

        for reservation in self._reservations:
            if (
                reservation.guest.id_number == guest.id_number
                and reservation.room.room_number == room.room_number
                and reservation.status in [Reservation.PENDING, Reservation.ACTIVE]
            ):
                raise DuplicateReservationError(
                    f"Guest {guest.id_number} already has an active reservation "
                    f"for room {room.room_number}."
                )

        reservation = Reservation(guest, room, nights)
        self._reservations.append(reservation)

        return reservation

    def cancel_reservation(self, reservation_id):
        """Cancel a reservation."""
        reservation = self.get_reservation_by_id(reservation_id)
        reservation.cancel()
        return reservation

    def check_in(self, reservation_id):
        """Check a guest in."""
        reservation = self.get_reservation_by_id(reservation_id)
        reservation.check_in()
        return reservation

    def check_out(self, reservation_id):
        """Check a guest out."""
        reservation = self.get_reservation_by_id(reservation_id)
        reservation.check_out()
        return reservation

    def add_service_to_reservation(self, reservation_id, service_id):
        """Add a service to a reservation."""
        reservation = self.get_reservation_by_id(reservation_id)
        service = self.get_service_by_id(service_id)

        reservation.add_service(service)
        return reservation

    def calculate_accommodation_cost(self, reservation_id):
        """Calculate accommodation cost only."""
        reservation = self.get_reservation_by_id(reservation_id)
        return reservation.calculate_room_charge()

    def calculate_total_charge(self, reservation_id):
        """Calculate total bill including services."""
        reservation = self.get_reservation_by_id(reservation_id)
        return reservation.calculate_total_charge()

    def generate_final_bill(self, reservation_id):
        """Generate final bill summary."""
        reservation = self.get_reservation_by_id(reservation_id)
        return reservation.get_bill_summary()

    def process_payment(self, reservation_id, payment_method, amount=None, **payment_details):
        """
        Process a payment for a reservation.

        Supported payment methods:
        - cash
        - card
        - mobile
        """
        reservation = self.get_reservation_by_id(reservation_id)

        if reservation.status == Reservation.CANCELLED:
            raise InvalidReservationStatusError(
                "Payments cannot be processed for a cancelled reservation."
            )

        if reservation.is_paid:
            raise PaymentError("Reservation is already fully paid.")

        if amount is None:
            amount = reservation.balance_due

        if isinstance(amount, bool) or not isinstance(amount, (int, float)):
            raise PaymentError("Payment amount must be a number.")

        if amount <= 0:
            raise PaymentError("Payment amount must be greater than zero.")

        if amount > reservation.balance_due + 0.01:
            raise PaymentError(
                "Payment amount cannot be greater than the outstanding balance."
            )

        payment_method = str(payment_method).strip().lower()

        if payment_method == "cash":
            payment = CashPayment(
                amount=amount,
                cash_received=payment_details.get("cash_received", amount)
            )

        elif payment_method == "card":
            payment = CardPayment(
                amount=amount,
                card_number=payment_details.get("card_number"),
                expiry_date=payment_details.get("expiry_date"),
                cvv=payment_details.get("cvv")
            )

        elif payment_method == "mobile":
            payment = MobilePayment(
                amount=amount,
                phone_number=payment_details.get("phone_number"),
                provider=payment_details.get("provider")
            )

        else:
            raise PaymentError(
                "Unsupported payment method. Use 'cash', 'card', or 'mobile'."
            )

        payment.process_payment()
        reservation.record_payment(payment)

        return payment

    def get_payment_history(self, reservation_id):
        """Return all payments made for a reservation."""
        reservation = self.get_reservation_by_id(reservation_id)
        return reservation.payments

    def generate_payment_receipt(self, payment):
        """Generate a receipt for a payment object."""
        if not isinstance(payment, Payment):
            raise TypeError("payment must be an instance of Payment.")

        return payment.get_receipt()

    def get_guest_history(self, guest_id):
        """Return all reservations belonging to a guest."""
        guest = self.get_guest_by_id(guest_id)

        history = [
            reservation for reservation in self._reservations
            if reservation.guest.id_number == guest.id_number
        ]

        return history

    def display_guest_history(self, guest_id):
        """Display all reservations belonging to a guest."""
        history = self.get_guest_history(guest_id)

        if not history:
            return "No reservation history found for this guest."

        return "\n".join(str(reservation) for reservation in history)

    def display_available_rooms(self):
        """Display all available rooms."""
        available_rooms = [
            str(room) for room in self._rooms
            if room.is_available()
        ]

        if not available_rooms:
            return "No available rooms."

        return "\n".join(available_rooms)

    def display_hotel_summary(self):
        """Display hotel summary including revenue information."""
        total_rooms = len(self._rooms)
        available_rooms = len([
            room for room in self._rooms
            if room.is_available()
        ])

        total_guests = len(self._guests)
        total_staff = len(self._staff)
        total_reservations = len(self._reservations)

        active_reservations = len([
            reservation for reservation in self._reservations
            if reservation.status in [Reservation.PENDING, Reservation.ACTIVE]
        ])

        completed_reservations = len([
            reservation for reservation in self._reservations
            if reservation.status == Reservation.COMPLETED
        ])

        cancelled_reservations = len([
            reservation for reservation in self._reservations
            if reservation.status == Reservation.CANCELLED
        ])

        total_revenue = round(sum(
            reservation.amount_paid for reservation in self._reservations
            if reservation.status != Reservation.CANCELLED
        ), 2)

        outstanding_balance = round(sum(
            reservation.balance_due for reservation in self._reservations
            if reservation.status != Reservation.CANCELLED
        ), 2)

        lines = []
        lines.append("=" * 45)
        lines.append(f"HOTEL SUMMARY - {self.name}")
        lines.append("=" * 45)
        lines.append(f"Total Rooms: {total_rooms}")
        lines.append(f"Available Rooms: {available_rooms}")
        lines.append(f"Total Guests: {total_guests}")
        lines.append(f"Total Staff: {total_staff}")
        lines.append(f"Total Reservations: {total_reservations}")
        lines.append(f"Active/Pending Reservations: {active_reservations}")
        lines.append(f"Completed Reservations: {completed_reservations}")
        lines.append(f"Cancelled Reservations: {cancelled_reservations}")
        lines.append(f"Total Revenue: {total_revenue:.2f}")
        lines.append(f"Outstanding Balance: {outstanding_balance:.2f}")
        lines.append("=" * 45)

        return "\n".join(lines)

    def __str__(self):
        return (
            f"Hotel: {self.name} | Rooms: {len(self._rooms)} | "
            f"Guests: {len(self._guests)}"
        )
````

#####  Object-to-Dictionary Converters

````python
def room_to_dict(room):
    """Convert a Room object into a dictionary."""
    room_class = room.__class__.__name__

    data = {
        "class": room_class,
        "room_number": room.room_number,
        "base_price": room.base_price,
        "status": room.status
    }

    if isinstance(room, StandardRoom):
        data["has_tv"] = room.has_tv

    elif isinstance(room, DeluxeRoom):
        data["includes_breakfast"] = room.includes_breakfast

    elif isinstance(room, Suite):
        data["lounge_access"] = room.lounge_access

    return data


def service_to_dict(service):
    """Convert a HotelService object into a dictionary."""
    return {
        "service_id": service.service_id,
        "name": service.name,
        "price": service.price
    }


def payment_to_dict(payment):
    """Convert a Payment object into a dictionary."""
    data = {
        "class": payment.__class__.__name__,
        "payment_id": payment.payment_id,
        "amount": payment.amount,
        "payment_date": payment.payment_date.isoformat(),
        "status": payment.status
    }

    if isinstance(payment, CashPayment):
        data["cash_received"] = payment.cash_received

    elif isinstance(payment, CardPayment):
        data["card_last4"] = payment.masked_card_number[-4:]
        data["expiry_date"] = payment._expiry_date

    elif isinstance(payment, MobilePayment):
        data["phone_number"] = payment.phone_number
        data["provider"] = payment.provider

    return data


def reservation_to_dict(reservation):
    """Convert a Reservation object into a dictionary."""
    return {
        "reservation_id": reservation.reservation_id,
        "guest_id": reservation.guest.id_number,
        "room_number": reservation.room.room_number,
        "nights": reservation.nights,
        "check_in_date": reservation.check_in_date.isoformat(),
        "status": reservation.status,
        "payment_status": reservation.payment_status,
        "created_on": reservation.created_on.isoformat(),
        "services": [
            service.service_id for service in reservation.services
        ],
        "payments": [
            payment_to_dict(payment) for payment in reservation.payments
        ]
    }


def guest_to_dict(guest):
    """Convert a Guest object into a dictionary."""
    return {
        "name": guest.name,
        "id_number": guest.id_number,
        "phone": guest.phone,
        "guest_type": guest.guest_type,
        "registered_on": guest.registered_on.isoformat(),
        "loyalty_points": guest.loyalty_points
    }


def staff_to_dict(staff):
    """Convert a Staff object into a dictionary."""
    return {
        "name": staff.name,
        "id_number": staff.id_number,
        "role": staff.role
    }


def hotel_to_dict(hotel):
    """Convert the whole Hotel object into a dictionary."""
    return {
        "hotel": {
            "name": hotel.name
        },
        "guests": [
            guest_to_dict(guest) for guest in hotel.guests
        ],
        "staff": [
            staff_to_dict(staff) for staff in hotel.staff
        ],
        "rooms": [
            room_to_dict(room) for room in hotel.rooms
        ],
        "services": [
            service_to_dict(service) for service in hotel.services
        ],
        "reservations": [
            reservation_to_dict(reservation)
            for reservation in hotel.reservations
        ]
    }
````

````python
def save_hotel(hotel, filepath=DATA_FILE):
    """
    Save the hotel system to a JSON file.
    """
    directory = os.path.dirname(filepath)

    if directory:
        os.makedirs(directory, exist_ok=True)

    hotel_data = hotel_to_dict(hotel)

    with open(filepath, "w") as file:
        json.dump(hotel_data, file, indent=4)

    return filepath
````

##### Dictionary-to-Object Converters

````python
def payment_from_dict(payment_data):
    """Recreate a Payment object from a dictionary."""
    payment_class = payment_data.get("class")

    if payment_class == "CashPayment":
        payment = CashPayment.__new__(CashPayment)

        payment._payment_id = payment_data["payment_id"]
        payment._amount = float(payment_data["amount"])
        payment._payment_date = datetime.fromisoformat(payment_data["payment_date"])
        payment._status = payment_data["status"]
        payment._cash_received = float(payment_data.get("cash_received", payment_data["amount"]))

        return payment

    elif payment_class == "CardPayment":
        payment = CardPayment.__new__(CardPayment)

        payment._payment_id = payment_data["payment_id"]
        payment._amount = float(payment_data["amount"])
        payment._payment_date = datetime.fromisoformat(payment_data["payment_date"])
        payment._status = payment_data["status"]

        card_last4 = payment_data.get("card_last4", "0000")
        payment._card_number = ("0000000000000000" + card_last4)[-16:]
        payment._expiry_date = payment_data.get("expiry_date", "12/99")
        payment._cvv = "000"

        return payment

    elif payment_class == "MobilePayment":
        payment = MobilePayment.__new__(MobilePayment)

        payment._payment_id = payment_data["payment_id"]
        payment._amount = float(payment_data["amount"])
        payment._payment_date = datetime.fromisoformat(payment_data["payment_date"])
        payment._status = payment_data["status"]
        payment._phone_number = payment_data.get("phone_number", "")
        payment._provider = payment_data.get("provider", "")

        return payment

    else:
        raise ValueError(f"Unknown payment class: {payment_class}")


def reservation_from_dict(hotel, reservation_data, guest_map, room_map, service_map):
    """Recreate a Reservation object from a dictionary."""
    guest = guest_map[reservation_data["guest_id"]]
    room = room_map[reservation_data["room_number"]]

    reservation = Reservation.__new__(Reservation)

    reservation._reservation_id = reservation_data["reservation_id"]
    reservation._guest = guest
    reservation._room = room
    reservation._nights = reservation_data["nights"]
    reservation._check_in_date = datetime.fromisoformat(
        reservation_data["check_in_date"]
    ).date()
    reservation._status = reservation_data["status"]
    reservation._payment_status = reservation_data["payment_status"]
    reservation._created_on = datetime.fromisoformat(reservation_data["created_on"])

    reservation._services = []

    for service_id in reservation_data.get("services", []):
        if service_id in service_map:
            reservation._services.append(service_map[service_id])

    reservation._payments = []

    for payment_data in reservation_data.get("payments", []):
        payment = payment_from_dict(payment_data)
        reservation._payments.append(payment)

    return reservation
````

##### Load Hotel Function

````python
def update_next_id(object_ids, prefix, target_class, attribute_name):
    """
    Update class ID counters after loading data so that new objects
    do not reuse old IDs.
    """
    maximum_value = 0

    for object_id in object_ids:
        try:
            number_part = str(object_id).split(prefix)[1]
            value = int(number_part)

            if value > maximum_value:
                maximum_value = value

        except Exception:
            continue

    if maximum_value > 0:
        setattr(target_class, attribute_name, maximum_value)


def load_hotel(filepath=DATA_FILE):
    """
    Load the hotel system from a JSON file.
    """
    if not os.path.exists(filepath):
        raise FileNotFoundError(f"Data file not found: {filepath}")

    with open(filepath, "r") as file:
        data = json.load(file)

    hotel = Hotel(data["hotel"]["name"])

    guest_map = {}
    room_map = {}
    service_map = {}

    # Load guests
    for guest_data in data.get("guests", []):
        guest = Guest(
            name=guest_data["name"],
            id_number=guest_data["id_number"],
            phone=guest_data["phone"],
            guest_type=guest_data.get("guest_type", "Regular")
        )

        guest.registered_on = datetime.fromisoformat(
            guest_data["registered_on"]
        ).date()

        guest._loyalty_points = guest_data.get("loyalty_points", 0)

        hotel._guests.append(guest)
        guest_map[guest.id_number] = guest

    Guest.guest_count = len(hotel._guests)

    # Load staff
    for staff_data in data.get("staff", []):
        staff = Staff(
            name=staff_data["name"],
            id_number=staff_data["id_number"],
            role=staff_data["role"]
        )

        hotel._staff.append(staff)

    # Load rooms
    for room_data in data.get("rooms", []):
        room_class = room_data.get("class")

        if room_class == "StandardRoom":
            room = StandardRoom(
                room_number=room_data["room_number"],
                base_price=room_data["base_price"],
                has_tv=room_data.get("has_tv", True)
            )

        elif room_class == "DeluxeRoom":
            room = DeluxeRoom(
                room_number=room_data["room_number"],
                base_price=room_data["base_price"],
                includes_breakfast=room_data.get("includes_breakfast", True)
            )

        elif room_class == "Suite":
            room = Suite(
                room_number=room_data["room_number"],
                base_price=room_data["base_price"],
                lounge_access=room_data.get("lounge_access", True)
            )

        else:
            raise ValueError(f"Unknown room class: {room_class}")

        room._status = room_data.get("status", Room.AVAILABLE)

        hotel._rooms.append(room)
        room_map[room.room_number] = room

    # Load services
    for service_data in data.get("services", []):
        service = HotelService(
            name=service_data["name"],
            price=service_data["price"]
        )

        service.service_id = service_data["service_id"]

        hotel._services.append(service)
        service_map[service.service_id] = service

    # Load reservations
    for reservation_data in data.get("reservations", []):
        reservation = reservation_from_dict(
            hotel=hotel,
            reservation_data=reservation_data,
            guest_map=guest_map,
            room_map=room_map,
            service_map=service_map
        )

        hotel._reservations.append(reservation)

    # Update ID counters
    update_next_id(
        object_ids=[service.service_id for service in hotel.services],
        prefix="SRV-",
        target_class=HotelService,
        attribute_name="_next_service_id"
    )

    update_next_id(
        object_ids=[
            payment.payment_id
            for reservation in hotel.reservations
            for payment in reservation.payments
        ],
        prefix="PAY-",
        target_class=Payment,
        attribute_name="_next_payment_id"
    )

    update_next_id(
        object_ids=[
            reservation.reservation_id
            for reservation in hotel.reservations
        ],
        prefix="RES-",
        target_class=Reservation,
        attribute_name="_next_reservation_id"
    )

    return hotel
````

````python
# CLI Helper Functions

def pause():
    """Pause the CLI until the user presses Enter."""
    input("\nPress Enter to continue...")


def read_positive_int(prompt):
    """Read a positive integer from the user."""
    value = input(prompt).strip()

    if not value.isdigit():
        raise ValueError("Please enter a positive integer.")

    number = int(value)

    if number <= 0:
        raise ValueError("Please enter a positive integer.")

    return number


def read_positive_float(prompt):
    """Read a positive number from the user."""
    value = input(prompt).strip()

    try:
        number = float(value)
    except ValueError:
        raise ValueError("Please enter a valid number.")

    if number <= 0:
        raise ValueError("Please enter a number greater than zero.")

    return number


def read_optional_float(prompt):
    """Read an optional number from the user."""
    value = input(prompt).strip()

    if value == "":
        return None

    try:
        number = float(value)
    except ValueError:
        raise ValueError("Please enter a valid number or leave the field blank.")

    if number <= 0:
        raise ValueError("Amount must be greater than zero.")

    return number
````

````python
def run_hotel_cli(cli_hotel):
    """
    Run the interactive command-line interface for the hotel system.
    """

    while True:
        print("\n")
        print("=" * 45)
        print(f"{cli_hotel.name.upper()} - HOTEL MANAGEMENT SYSTEM")
        print("=" * 45)
        print("1. Register guest")
        print("2. Register staff")
        print("3. Add room")
        print("4. Add service")
        print("5. View available rooms")
        print("6. Search rooms")
        print("7. Make reservation")
        print("8. Cancel reservation")
        print("9. Check in")
        print("10. Check out")
        print("11. Add service to reservation")
        print("12. Generate final bill")
        print("13. Process payment")
        print("14. View payment history")
        print("15. View guest history")
        print("16. Display hotel summary")
        print("17. Save hotel data")
        print("18. Exit")
        print("=" * 45)

        choice = input("Enter your choice: ").strip()

        try:
            if choice == "1":
                name = input("Guest name: ").strip()
                id_number = input("Guest ID: ").strip()
                phone = input("Phone number: ").strip()
                guest_type = input("Guest type (Regular/VIP): ").strip()

                if guest_type == "":
                    guest_type = "Regular"

                guest = cli_hotel.register_guest(
                    name=name,
                    id_number=id_number,
                    phone=phone,
                    guest_type=guest_type
                )

                print("\nGuest registered successfully:")
                print(guest)

            elif choice == "2":
                name = input("Staff name: ").strip()
                id_number = input("Staff ID: ").strip()
                role = input("Staff role: ").strip()

                staff = cli_hotel.register_staff(
                    name=name,
                    id_number=id_number,
                    role=role
                )

                print("\nStaff registered successfully:")
                print(staff)

            elif choice == "3":
                print("Room types: standard, deluxe, suite")
                room_type = input("Room type: ").strip()
                room_number = input("Room number: ").strip()
                base_price = read_positive_float("Base price: ")

                room = cli_hotel.add_room(
                    room_type=room_type,
                    room_number=room_number,
                    base_price=base_price
                )

                print("\nRoom added successfully:")
                print(room)

            elif choice == "4":
                service_name = input("Service name: ").strip()
                service_price = read_positive_float("Service price: ")

                service = cli_hotel.add_service(
                    name=service_name,
                    price=service_price
                )

                print("\nService added successfully:")
                print(service)

            elif choice == "5":
                print("\nAvailable Rooms:")
                print(cli_hotel.display_available_rooms())

            elif choice == "6":
                room_number = input("Room number (leave blank to skip): ").strip()
                room_type = input("Room type (standard/deluxe/suite, leave blank to skip): ").strip()
                available_only_input = input("Show available only? y/n: ").strip().lower()

                available_only = available_only_input == "y"

                if room_number == "":
                    room_number = None

                if room_type == "":
                    room_type = None

                results = cli_hotel.search_rooms(
                    room_number=room_number,
                    room_type=room_type,
                    available_only=available_only
                )

                if results:
                    print("\nSearch Results:")
                    for room in results:
                        print(room)
                else:
                    print("\nNo rooms found.")

            elif choice == "7":
                guest_id = input("Guest ID: ").strip()
                room_number = input("Room number: ").strip()
                nights = read_positive_int("Number of nights: ")

                reservation = cli_hotel.make_reservation(
                    guest_id=guest_id,
                    room_number=room_number,
                    nights=nights
                )

                print("\nReservation created successfully:")
                print(reservation)

            elif choice == "8":
                reservation_id = input("Reservation ID: ").strip()

                reservation = cli_hotel.cancel_reservation(reservation_id)

                print("\nReservation cancelled successfully:")
                print(reservation)

            elif choice == "9":
                reservation_id = input("Reservation ID: ").strip()

                reservation = cli_hotel.check_in(reservation_id)

                print("\nCheck-in successful:")
                print(reservation)

            elif choice == "10":
                reservation_id = input("Reservation ID: ").strip()

                reservation = cli_hotel.check_out(reservation_id)

                print("\nCheck-out successful:")
                print(reservation)

            elif choice == "11":
                print("\nAvailable Services:")
                for service in cli_hotel.services:
                    print(service)

                reservation_id = input("Reservation ID: ").strip()
                service_id = input("Service ID: ").strip()

                reservation = cli_hotel.add_service_to_reservation(
                    reservation_id=reservation_id,
                    service_id=service_id
                )

                print("\nService added successfully.")
                print(reservation)

            elif choice == "12":
                reservation_id = input("Reservation ID: ").strip()

                print()
                print(cli_hotel.generate_final_bill(reservation_id))

            elif choice == "13":
                reservation_id = input("Reservation ID: ").strip()

                print("Payment methods: cash, card, mobile")
                payment_method = input("Payment method: ").strip()

                amount = read_optional_float(
                    "Amount (leave blank to pay full balance): "
                )

                if payment_method.lower() == "cash":
                    cash_received = read_optional_float(
                        "Cash received (leave blank for exact amount): "
                    )

                    payment = cli_hotel.process_payment(
                        reservation_id=reservation_id,
                        payment_method=payment_method,
                        amount=amount,
                        cash_received=cash_received
                    )

                elif payment_method.lower() == "card":
                    card_number = input("Card number: ").strip()
                    expiry_date = input("Expiry date MM/YY: ").strip()
                    cvv = input("CVV: ").strip()

                    payment = cli_hotel.process_payment(
                        reservation_id=reservation_id,
                        payment_method=payment_method,
                        amount=amount,
                        card_number=card_number,
                        expiry_date=expiry_date,
                        cvv=cvv
                    )

                elif payment_method.lower() == "mobile":
                    phone_number = input("Phone number: ").strip()
                    provider = input("Provider: ").strip()

                    payment = cli_hotel.process_payment(
                        reservation_id=reservation_id,
                        payment_method=payment_method,
                        amount=amount,
                        phone_number=phone_number,
                        provider=provider
                    )

                else:
                    raise PaymentError(
                        "Unsupported payment method. Use cash, card, or mobile."
                    )

                print("\nPayment processed successfully.")
                print(cli_hotel.generate_payment_receipt(payment))

            elif choice == "14":
                reservation_id = input("Reservation ID: ").strip()

                payments = cli_hotel.get_payment_history(reservation_id)

                if payments:
                    print("\nPayment History:")
                    for payment in payments:
                        print(payment)
                else:
                    print("\nNo payments found for this reservation.")

            elif choice == "15":
                guest_id = input("Guest ID: ").strip()

                print()
                print(cli_hotel.display_guest_history(guest_id))

            elif choice == "16":
                print()
                print(cli_hotel.display_hotel_summary())

            elif choice == "17":
                file_path = save_hotel(cli_hotel)
                print(f"\nHotel data saved successfully to {file_path}")

            elif choice == "18":
                print("\nThank you for using the Hotel Management System.")
                break

            else:
                print("\nInvalid choice. Please enter a number from 1 to 18.")

        except HotelError as error:
            print(f"\nHotel error: {error}")

        except ValueError as error:
            print(f"\nInput error: {error}")

        except Exception as error:
            print(f"\nUnexpected error: {error}")

        pause()
````

#### USER INTERFACE USING GUI

##### Graphical User Interface (Tkinter)

This section provides a desktop Graphical User Interface for the hotel system using Python's built-in `tkinter` library. 

The GUI allows users to interact with the `Hotel` controller object visually through tabs for the Dashboard, Rooms, and Reservations.

````python
import tkinter as tk
from tkinter import ttk, messagebox, simpledialog


class HotelGUI:
    """
    Graphical User Interface for the Hotel Reservation and Management System.

    This GUI uses the Hotel controller class and does not contain
    the main business logic. All operations are delegated to the Hotel object.
    """

    def __init__(self, hotel):
        self.hotel = hotel

        self.root = tk.Tk()
        self.root.title(f"{self.hotel.name} - Hotel Management System")
        self.root.geometry("1150x700")

        self.create_layout()
        self.refresh_all()

    def run(self):
        """Start the GUI main loop."""
        self.root.mainloop()

    def create_layout(self):
        """Create notebook tabs."""
        self.notebook = ttk.Notebook(self.root)
        self.notebook.pack(fill="both", expand=True, padx=10, pady=10)

        self.dashboard_tab = ttk.Frame(self.notebook)
        self.rooms_tab = ttk.Frame(self.notebook)
        self.guests_tab = ttk.Frame(self.notebook)
        self.reservations_tab = ttk.Frame(self.notebook)

        self.notebook.add(self.dashboard_tab, text="Dashboard")
        self.notebook.add(self.rooms_tab, text="Rooms")
        self.notebook.add(self.guests_tab, text="Guests")
        self.notebook.add(self.reservations_tab, text="Reservations & Payments")

        self.build_dashboard()
        self.build_rooms()
        self.build_guests()
        self.build_reservations()

    def refresh_all(self):
        """Refresh all GUI sections."""
        self.refresh_dashboard()
        self.refresh_rooms()
        self.refresh_guests()
        self.refresh_reservations()

    # Dashboard Tab

    def build_dashboard(self):
        """Build dashboard tab."""
        button_frame = ttk.Frame(self.dashboard_tab)
        button_frame.pack(fill="x", pady=5)

        ttk.Button(
            button_frame,
            text="Refresh Dashboard",
            command=self.refresh_dashboard
        ).pack(side="left", padx=5)

        self.dashboard_text = tk.Text(self.dashboard_tab)
        self.dashboard_text.pack(fill="both", expand=True, padx=5, pady=5)

    def refresh_dashboard(self):
        """Refresh dashboard summary."""
        self.dashboard_text.delete("1.0", tk.END)

        content = self.hotel.display_hotel_summary()
        content += "\n\nAVAILABLE ROOMS\n"
        content += self.hotel.display_available_rooms()

        self.dashboard_text.insert(tk.END, content)

    # Rooms Tab

    def build_rooms(self):
        """Build rooms tab."""
        button_frame = ttk.Frame(self.rooms_tab)
        button_frame.pack(fill="x", pady=5)

        ttk.Button(
            button_frame,
            text="Add Room",
            command=self.add_room_dialog
        ).pack(side="left", padx=5)

        ttk.Button(
            button_frame,
            text="Refresh Rooms",
            command=self.refresh_rooms
        ).pack(side="left", padx=5)

        columns = ("room_number", "room_type", "status", "base_price")

        self.rooms_tree = ttk.Treeview(
            self.rooms_tab,
            columns=columns,
            show="headings"
        )

        self.rooms_tree.heading("room_number", text="Room Number")
        self.rooms_tree.heading("room_type", text="Room Type")
        self.rooms_tree.heading("status", text="Status")
        self.rooms_tree.heading("base_price", text="Base Price")

        self.rooms_tree.column("room_number", width=150)
        self.rooms_tree.column("room_type", width=250)
        self.rooms_tree.column("status", width=150)
        self.rooms_tree.column("base_price", width=150)

        self.rooms_tree.pack(fill="both", expand=True, padx=5, pady=5)

    def refresh_rooms(self):
        """Refresh room table."""
        for item in self.rooms_tree.get_children():
            self.rooms_tree.delete(item)

        for room in self.hotel.rooms:
            self.rooms_tree.insert(
                "",
                tk.END,
                values=(
                    room.room_number,
                    room.__class__.__name__,
                    room.status,
                    f"{room.base_price:.2f}"
                )
            )

    def add_room_dialog(self):
        """Open dialog to add a room."""
        try:
            room_type = simpledialog.askstring(
                "Add Room",
                "Room type (standard/deluxe/suite):",
                parent=self.root
            )

            if not room_type:
                return

            room_number = simpledialog.askstring(
                "Add Room",
                "Room number:",
                parent=self.root
            )

            if not room_number:
                return

            base_price = simpledialog.askfloat(
                "Add Room",
                "Base price:",
                parent=self.root
            )

            if not base_price:
                return

            self.hotel.add_room(
                room_type=room_type,
                room_number=room_number,
                base_price=base_price
            )

            self.refresh_rooms()
            self.refresh_dashboard()

            messagebox.showinfo(
                "Success",
                "Room added successfully."
            )

        except Exception as error:
            messagebox.showerror("Error", str(error))

    # Guests Tab

    def build_guests(self):
        """Build guests tab."""
        button_frame = ttk.Frame(self.guests_tab)
        button_frame.pack(fill="x", pady=5)

        ttk.Button(
            button_frame,
            text="Register Guest",
            command=self.register_guest_dialog
        ).pack(side="left", padx=5)

        ttk.Button(
            button_frame,
            text="Refresh Guests",
            command=self.refresh_guests
        ).pack(side="left", padx=5)

        columns = ("id_number", "name", "phone", "guest_type")

        self.guests_tree = ttk.Treeview(
            self.guests_tab,
            columns=columns,
            show="headings"
        )

        self.guests_tree.heading("id_number", text="Guest ID")
        self.guests_tree.heading("name", text="Name")
        self.guests_tree.heading("phone", text="Phone")
        self.guests_tree.heading("guest_type", text="Guest Type")

        self.guests_tree.column("id_number", width=180)
        self.guests_tree.column("name", width=250)
        self.guests_tree.column("phone", width=200)
        self.guests_tree.column("guest_type", width=150)

        self.guests_tree.pack(fill="both", expand=True, padx=5, pady=5)

    def refresh_guests(self):
        """Refresh guest table."""
        for item in self.guests_tree.get_children():
            self.guests_tree.delete(item)

        for guest in self.hotel.guests:
            self.guests_tree.insert(
                "",
                tk.END,
                values=(
                    guest.id_number,
                    guest.name,
                    guest.phone,
                    guest.guest_type
                )
            )

    def register_guest_dialog(self):
        """Open dialog to register a guest."""
        try:
            name = simpledialog.askstring(
                "Register Guest",
                "Guest name:",
                parent=self.root
            )

            if not name:
                return

            id_number = simpledialog.askstring(
                "Register Guest",
                "Guest ID:",
                parent=self.root
            )

            if not id_number:
                return

            phone = simpledialog.askstring(
                "Register Guest",
                "Phone number:",
                parent=self.root
            )

            if not phone:
                return

            guest_type = simpledialog.askstring(
                "Register Guest",
                "Guest type (Regular/VIP):",
                parent=self.root
            )

            if not guest_type:
                guest_type = "Regular"

            self.hotel.register_guest(
                name=name,
                id_number=id_number,
                phone=phone,
                guest_type=guest_type
            )

            self.refresh_guests()
            self.refresh_dashboard()

            messagebox.showinfo(
                "Success",
                "Guest registered successfully."
            )

        except Exception as error:
            messagebox.showerror("Error", str(error))


    # Reservations Tab

    def build_reservations(self):
        """Build reservations and payments tab."""
        button_frame = ttk.Frame(self.reservations_tab)
        button_frame.pack(fill="x", pady=5)

        ttk.Button(
            button_frame,
            text="Make Reservation",
            command=self.make_reservation_dialog
        ).pack(side="left", padx=5)

        ttk.Button(
            button_frame,
            text="Cancel Reservation",
            command=self.cancel_reservation_action
        ).pack(side="left", padx=5)

        ttk.Button(
            button_frame,
            text="Check In",
            command=self.check_in_action
        ).pack(side="left", padx=5)

        ttk.Button(
            button_frame,
            text="Check Out",
            command=self.check_out_action
        ).pack(side="left", padx=5)

        ttk.Button(
            button_frame,
            text="Final Bill",
            command=self.show_final_bill
        ).pack(side="left", padx=5)

        ttk.Button(
            button_frame,
            text="Process Payment",
            command=self.process_payment_dialog
        ).pack(side="left", padx=5)

        ttk.Button(
            button_frame,
            text="Refresh",
            command=self.refresh_reservations
        ).pack(side="left", padx=5)

        columns = (
            "reservation_id",
            "guest",
            "room",
            "nights",
            "status",
            "payment_status",
            "total"
        )

        self.reservations_tree = ttk.Treeview(
            self.reservations_tab,
            columns=columns,
            show="headings"
        )

        self.reservations_tree.heading("reservation_id", text="Reservation ID")
        self.reservations_tree.heading("guest", text="Guest")
        self.reservations_tree.heading("room", text="Room")
        self.reservations_tree.heading("nights", text="Nights")
        self.reservations_tree.heading("status", text="Status")
        self.reservations_tree.heading("payment_status", text="Payment Status")
        self.reservations_tree.heading("total", text="Total")

        self.reservations_tree.column("reservation_id", width=130)
        self.reservations_tree.column("guest", width=200)
        self.reservations_tree.column("room", width=100)
        self.reservations_tree.column("nights", width=80)
        self.reservations_tree.column("status", width=120)
        self.reservations_tree.column("payment_status", width=150)
        self.reservations_tree.column("total", width=120)

        self.reservations_tree.pack(fill="both", expand=True, padx=5, pady=5)

    def refresh_reservations(self):
        """Refresh reservation table."""
        for item in self.reservations_tree.get_children():
            self.reservations_tree.delete(item)

        for reservation in self.hotel.reservations:
            self.reservations_tree.insert(
                "",
                tk.END,
                values=(
                    reservation.reservation_id,
                    reservation.guest.name,
                    reservation.room.room_number,
                    reservation.nights,
                    reservation.status,
                    reservation.payment_status,
                    f"{reservation.calculate_total_charge():.2f}"
                )
            )

    def get_selected_reservation_id(self):
        """Get selected reservation ID from table."""
        selected = self.reservations_tree.selection()

        if not selected:
            messagebox.showwarning(
                "No Selection",
                "Please select a reservation first."
            )
            return None

        item = self.reservations_tree.item(selected[0])
        reservation_id = item["values"][0]

        return str(reservation_id)

    def make_reservation_dialog(self):
        """Open dialog to make a reservation."""
        try:
            guest_id = simpledialog.askstring(
                "Make Reservation",
                "Guest ID:",
                parent=self.root
            )

            if not guest_id:
                return

            room_number = simpledialog.askstring(
                "Make Reservation",
                "Room number:",
                parent=self.root
            )

            if not room_number:
                return

            nights = simpledialog.askinteger(
                "Make Reservation",
                "Number of nights:",
                parent=self.root
            )

            if not nights:
                return

            self.hotel.make_reservation(
                guest_id=guest_id,
                room_number=room_number,
                nights=nights
            )

            self.refresh_reservations()
            self.refresh_rooms()
            self.refresh_dashboard()

            messagebox.showinfo(
                "Success",
                "Reservation created successfully."
            )

        except Exception as error:
            messagebox.showerror("Error", str(error))

    def cancel_reservation_action(self):
        """Cancel selected reservation."""
        reservation_id = self.get_selected_reservation_id()

        if not reservation_id:
            return

        try:
            self.hotel.cancel_reservation(reservation_id)

            self.refresh_reservations()
            self.refresh_rooms()
            self.refresh_dashboard()

            messagebox.showinfo(
                "Success",
                "Reservation cancelled successfully."
            )

        except Exception as error:
            messagebox.showerror("Error", str(error))

    def check_in_action(self):
        """Check in selected reservation."""
        reservation_id = self.get_selected_reservation_id()

        if not reservation_id:
            return

        try:
            self.hotel.check_in(reservation_id)

            self.refresh_reservations()
            self.refresh_rooms()
            self.refresh_dashboard()

            messagebox.showinfo(
                "Success",
                "Check-in successful."
            )

        except Exception as error:
            messagebox.showerror("Error", str(error))

    def check_out_action(self):
        """Check out selected reservation."""
        reservation_id = self.get_selected_reservation_id()

        if not reservation_id:
            return

        try:
            self.hotel.check_out(reservation_id)

            self.refresh_reservations()
            self.refresh_rooms()
            self.refresh_dashboard()

            messagebox.showinfo(
                "Success",
                "Check-out successful."
            )

        except Exception as error:
            messagebox.showerror("Error", str(error))

    def show_final_bill(self):
        """Show final bill for selected reservation."""
        reservation_id = self.get_selected_reservation_id()

        if not reservation_id:
            return

        try:
            bill = self.hotel.generate_final_bill(reservation_id)
            messagebox.showinfo("Final Bill", bill)

        except Exception as error:
            messagebox.showerror("Error", str(error))

    def process_payment_dialog(self):
        """Open dialog to process payment."""
        reservation_id = self.get_selected_reservation_id()

        if not reservation_id:
            return

        try:
            payment_method = simpledialog.askstring(
                "Process Payment",
                "Payment method (cash/card/mobile):",
                parent=self.root
            )

            if not payment_method:
                return

            amount_text = simpledialog.askstring(
                "Process Payment",
                "Amount (leave blank to pay full balance):",
                parent=self.root
            )

            if amount_text is None:
                return

            amount = float(amount_text) if amount_text.strip() != "" else None

            payment_details = {}

            if payment_method.lower() == "cash":
                cash_text = simpledialog.askstring(
                    "Cash Payment",
                    "Cash received (leave blank for exact amount):",
                    parent=self.root
                )

                if cash_text is None:
                    return

                if cash_text.strip() != "":
                    payment_details["cash_received"] = float(cash_text)

            elif payment_method.lower() == "card":
                card_number = simpledialog.askstring(
                    "Card Payment",
                    "Card number:",
                    parent=self.root
                )

                if card_number is None:
                    return

                expiry_date = simpledialog.askstring(
                    "Card Payment",
                    "Expiry date MM/YY:",
                    parent=self.root
                )

                if expiry_date is None:
                    return

                cvv = simpledialog.askstring(
                    "Card Payment",
                    "CVV:",
                    parent=self.root
                )

                if cvv is None:
                    return

                payment_details["card_number"] = card_number
                payment_details["expiry_date"] = expiry_date
                payment_details["cvv"] = cvv

            elif payment_method.lower() == "mobile":
                phone_number = simpledialog.askstring(
                    "Mobile Payment",
                    "Phone number:",
                    parent=self.root
                )

                if phone_number is None:
                    return

                provider = simpledialog.askstring(
                    "Mobile Payment",
                    "Provider:",
                    parent=self.root
                )

                if provider is None:
                    return

                payment_details["phone_number"] = phone_number
                payment_details["provider"] = provider

            else:
                raise PaymentError(
                    "Unsupported payment method. Use cash, card, or mobile."
                )

            payment = self.hotel.process_payment(
                reservation_id=reservation_id,
                payment_method=payment_method,
                amount=amount,
                **payment_details
            )

            self.refresh_reservations()
            self.refresh_dashboard()

            messagebox.showinfo(
                "Payment Successful",
                payment.get_receipt()
            )

        except Exception as error:
            messagebox.showerror("Error", str(error))
````

##### Deployment Plan

The deployed system uses:

- Backend API -> FastAPI 
- Frontend -> React + Vite 
- Frontend Hosting -> Vercel 
- Backend Hosting -> Render or Railway 
- Data Storage -> JSON for demonstration
- Repository -> GitHub

The deployed application is an extension used for demonstration.

````python
from fastapi import FastAPI, HTTPException
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
from typing import Optional

# ---------------------------------------------------------------
# 1. Initialize the FastAPI application
# ---------------------------------------------------------------
app = FastAPI(title="Hotel Management API")

# ---------------------------------------------------------------
# 2. Allow the React frontend to connect to this backend
# ---------------------------------------------------------------
app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"],
)

# ---------------------------------------------------------------
# 3. Pydantic models (validate incoming data from React)
# ---------------------------------------------------------------
class GuestInput(BaseModel):
    name: str
    id_number: str
    phone: str
    guest_type: str = "Regular"

class RoomInput(BaseModel):
    room_type: str
    room_number: str
    base_price: float

class ReservationInput(BaseModel):
    guest_id: str
    room_number: str
    nights: int

class PaymentInput(BaseModel):
    reservation_id: str
    payment_method: str
    amount: Optional[float] = None
    cash_received: Optional[float] = None
    card_number: Optional[str] = None
    expiry_date: Optional[str] = None
    cvv: Optional[str] = None
    phone_number: Optional[str] = None
    provider: Optional[str] = None

# ---------------------------------------------------------------
# 4. Helpers: convert OOP objects to JSON for React
# ---------------------------------------------------------------
def room_to_dict(room):
    return {
        "room_number": room.room_number,
        "status": room.status,
        "base_price": room.base_price,
        "description": room.get_description(),
    }

def reservation_to_dict(res):
    return {
        "reservation_id": res.reservation_id,
        "guest_name": res.guest.name,
        "room_number": res.room.room_number,
        "nights": res.nights,
        "status": res.status,
        "payment_status": res.payment_status,
        "total_charge": res.calculate_total_charge(),
        "amount_paid": res.amount_paid,
        "balance_due": res.balance_due,
    }

def payment_to_dict(p):
    return {
        "payment_id": p.payment_id,
        "method": p.get_payment_method(),
        "amount": p.amount,
        "status": p.status,
        "date": p.payment_date.strftime("%Y-%m-%d %H:%M"),
    }

def find_reservation(reservation_id):
    res = next((r for r in hotel.reservations if r.reservation_id == reservation_id), None)
    if res is None:
        raise HTTPException(status_code=404, detail=f"Reservation {reservation_id} not found")
    return res

# ---------------------------------------------------------------
# 5. API routes
# ---------------------------------------------------------------
@app.get("/")
def root():
    return {"message": "Hotel Management API is running"}

# ---- Dashboard ----
@app.get("/stats")
def get_stats():
    rooms = hotel.rooms
    live = [r for r in hotel.reservations if r.status != "CANCELLED"]
    return {
        "total_rooms": len(rooms),
        "available_rooms": len([r for r in rooms if r.is_available()]),
        "occupied_rooms": len([r for r in rooms if r.status == "OCCUPIED"]),
        "reserved_rooms": len([r for r in rooms if r.status == "RESERVED"]),
        "total_guests": len(hotel.guests),
        "total_reservations": len(hotel.reservations),
        "active_reservations": len([r for r in hotel.reservations if r.status in ("PENDING", "ACTIVE")]),
        "revenue": round(sum(r.amount_paid for r in live), 2),
        "outstanding": round(sum(r.balance_due for r in live), 2),
    }

@app.get("/summary")
def get_summary():
    summary = hotel.display_hotel_summary()
    if summary is None:      # fallback in case the method only prints
        summary = ""
    return {"summary": summary}

# ---- Rooms ----
@app.get("/rooms")
def get_rooms():
    # Uses the `hotel` object created earlier in your notebook
    return [room_to_dict(room) for room in hotel.rooms]

@app.get("/rooms/available")
def get_available_rooms():
    return [room_to_dict(room) for room in hotel.rooms if room.is_available()]

@app.post("/rooms")
def create_room(payload: RoomInput):
    try:
        room = hotel.add_room(payload.room_type, payload.room_number, payload.base_price)
        return {"message": "Room added successfully", "room": room_to_dict(room)}
    except Exception as e:
        raise HTTPException(status_code=400, detail=str(e))

# ---- Guests ----
@app.get("/guests")
def get_guests():
    return [
        {
            "id_number": g.id_number,
            "name": g.name,
            "phone": g.phone,
            "guest_type": g.guest_type,
            "registered_on": str(g.registered_on),
        }
        for g in hotel.guests
    ]

@app.post("/guests")
def create_guest(payload: GuestInput):
    try:
        guest = hotel.register_guest(payload.name, payload.id_number, payload.phone, payload.guest_type)
        return {"message": "Guest registered successfully", "guest": str(guest)}
    except Exception as e:
        raise HTTPException(status_code=400, detail=str(e))

# ---- Reservations ----
@app.get("/reservations")
def get_reservations():
    return [reservation_to_dict(res) for res in hotel.reservations]

@app.post("/reservations")
def create_reservation(payload: ReservationInput):
    try:
        res = hotel.make_reservation(payload.guest_id, payload.room_number, payload.nights)
        return {"message": "Reservation created successfully", "reservation": reservation_to_dict(res)}
    except Exception as e:
        raise HTTPException(status_code=400, detail=str(e))

@app.get("/reservations/{reservation_id}/bill")
def get_bill(reservation_id: str):
    try:
        return {"bill": hotel.generate_final_bill(reservation_id)}
    except Exception as e:
        raise HTTPException(status_code=400, detail=str(e))

@app.get("/reservations/{reservation_id}/payments")
def get_reservation_payments(reservation_id: str):
    res = find_reservation(reservation_id)
    return [payment_to_dict(p) for p in res.payments]

@app.post("/reservations/{reservation_id}/checkin")
def check_in(reservation_id: str):
    try:
        res = hotel.check_in(reservation_id)
        return {"message": f"{res.guest.name} checked into room {res.room.room_number}."}
    except Exception as e:
        raise HTTPException(status_code=400, detail=str(e))

@app.post("/reservations/{reservation_id}/checkout")
def check_out(reservation_id: str):
    try:
        res = hotel.check_out(reservation_id)
        return {"message": f"{res.guest.name} checked out of room {res.room.room_number}."}
    except Exception as e:
        raise HTTPException(status_code=400, detail=str(e))

@app.post("/reservations/{reservation_id}/cancel")
def cancel_reservation(reservation_id: str):
    try:
        hotel.cancel_reservation(reservation_id)
        return {"message": f"Reservation {reservation_id} cancelled."}
    except Exception as e:
        raise HTTPException(status_code=400, detail=str(e))

# ---- Payments ----
@app.post("/payments")
def process_payment(payload: PaymentInput):
    try:
        payment = hotel.process_payment(
            reservation_id=payload.reservation_id,
            payment_method=payload.payment_method,
            amount=payload.amount,
            cash_received=payload.cash_received,
            card_number=payload.card_number,
            expiry_date=payload.expiry_date,
            cvv=payload.cvv,
            phone_number=payload.phone_number,
            provider=payload.provider,
        )
        return {"message": "Payment processed successfully", "receipt": payment.get_receipt()}
    except Exception as e:
        raise HTTPException(status_code=400, detail=str(e))
````

##### Challenges Encountered



-  Challenge 1: Managing Room Availability
The system had to ensure that a reserved room could not be reserved again. This was solved by updating the room status immediately when a reservation object is created.

- Challenge 2: Preventing Duplicate Reservations
A guest could accidentally make multiple active reservations for the same room. This was solved by checking existing active or pending reservations before creating a new one.

- Challenge 3: Payment Tracking
The system needed to support partial payments and full payments. This was solved by adding payment tracking fields to the Reservation class and calculating balance due dynamically.

1. **Object Serialization:** Converting nested OOP objects (Reservations containing Payments) into JSON was challenging. Solved by writing custom `to_dict()` and `from_dict()` methods.
2. **State Management:** Ensuring room status updated immediately upon reservation to prevent double-booking. Solved by triggering `room.make_reserved()` inside the `Reservation` constructor.
3. **Polymorphic Billing:** Getting different room types to calculate totals uniformly. Solved by enforcing an abstract `calculate_cost()` method in the base `Room` class.
