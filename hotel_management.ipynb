#### 5. Class Design

<!-- ```text -->

The main classes in this project are:

- Person -> Abstract base class for people. Attributes include:`name`, `id_number`. Methods include: `get_details()`, `get_role()`. 
    - Guest ->  Represents a hotel guest. Attributes include: `phone`, `guest_type`.Methods include: `loyalty_points`, `add_loyalty_points()`.
    - Staff -> Represents hotel staff. Attributes include: `role`. Methods include: `get_role()`.
- Room -> Abstract base class for rooms. Attributes include: room_number, base_price, status. Methods include: `calculate_cost(), get_description()`.
    - StandardRoom -> Basic room. Attributes include: `has_tv` . Methods include: `calculate_cost()`.  
    - DeluxeRoom ->  Premium room. Attributes include: `includes_breakfast`. Methods include: `calculate_cost()`.
    - Suite -> Luxury room. Attributes include: `lounge_access`. Methods include: `calculate_cost()`. 
- Reservation -> Represents a guest booking. Attributes include: `guest, room, nights, services`. Methods include: `calculate_total_charge(), check_in(), check_out()`.
- HotelService ->  Extra hotel service. Attributes include: `service_id, name, price`. 
- Payment -> Base class for payment methods
- Hotel -> Central controller for guests, rooms, and reservations. Methods include: `make_reservation(), check_in(), generate_final_bill()`.

#### 6. OOP Concepts Applied


- Classes and Objects -> Classes such as Guest, Room, Reservation, and Hotel are used to create objects that represent real hotel entities.

- Encapsulation -> Sensitive or internal data is protected using private attributes such as `_status`, `_services`, `_guests`, and `_rooms`. Properties control access.

- Inheritance -> Guest and Staff inherit from Person. StandardRoom, DeluxeRoom, and Suite inherit from Room.

- Method Overriding -> Child classes override methods such as `get_role()`, `get_description()`, and `calculate_cost()`.

- Polymorphism -> Different room types calculate cost differently using the same method name: `calculate_cost(nights)`.

- Abstraction -> Person and Room are abstract base classes. They define abstract methods that child classes must implement.

- Composition -> A Reservation contains a Guest object, a Room object, and HotelService objects. The Hotel class contains collections of guests, rooms, reservations, and services.

- Class Methods -> `Guest.number_of_guests()` is a class method.

- Static Methods -> `Guest.validate_phone()` and `Room.validate_nights()` are static methods.

- Special Methods -> Special methods such as `__str__()`, `__repr__()`, `__len__()`, `__eq__()`, and `__hash__()` are implemented.

- Exception Handling -> Custom exceptions such as `RoomUnavailableError`, `GuestNotFoundError`, `InvalidNightsError`, and `InvalidReservationStatusError` are used.

````python
room_101 = hotel.add_room("standard", "101", 5000)
room_201 = hotel.add_room("deluxe", "201", 8000)
room_301 = hotel.add_room("suite", "301", 15000)

laundry = hotel.add_service("Laundry", 1500)
breakfast = hotel.add_service("Breakfast", 2000)

print("Rooms Added:")
print(room_101)
print(room_201)
print(room_301)

print("\nServices Added:")
print(laundry)
print(breakfast)
````

````python
print("Available Rooms:")
print(hotel.display_available_rooms())
````

````python
standard_rooms = hotel.search_rooms(room_type="standard", available_only=True)

print("Available Standard Rooms:")
for room in standard_rooms:
    print(room)
````

````python
reservation_1 = hotel.make_reservation("CU/2026/001", "101", 3)
reservation_2 = hotel.make_reservation("CU/2026/002", "301", 2)

print(reservation_1)
print(reservation_2)

print("\nAvailable Rooms After Reservations:")
print(hotel.display_available_rooms())
````

````python
hotel.add_service_to_reservation(reservation_1.reservation_id, laundry.service_id)
hotel.add_service_to_reservation(reservation_1.reservation_id, breakfast.service_id)

print(hotel.generate_final_bill(reservation_1.reservation_id))
````

````python
accommodation_cost = hotel.calculate_accommodation_cost(reservation_1.reservation_id)
total_charge = hotel.calculate_total_charge(reservation_1.reservation_id)

print(f"Accommodation Cost: {accommodation_cost}")
print(f"Total Charge: {total_charge}")
````

````python
hotel.check_in(reservation_1.reservation_id)
print(reservation_1)

hotel.check_out(reservation_1.reservation_id)
print(reservation_1)

print("\nAvailable Rooms After Check-Out:")
print(hotel.display_available_rooms())
````
