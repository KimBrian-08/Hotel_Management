##  OOP Concepts Applied

- Classes and Objects -> Classes such as Guest, Room, Reservation, and Hotel are used to create objects that represent real hotel entities.

- Encapsulation -> Sensitive or internal data is protected using private attributes such as `_status`, `_services`, `_guests`, and `_rooms`. Properties control access.

- Inheritance -> Guest and Staff inherit from Person. StandardRoom, DeluxeRoom, and Suite inherit from Room.

- Method Overriding -> Child classes override methods such as `get_role()`, `get_description()`, and `calculate_cost()`.

- Polymorphism -> Different room types calculate cost differently using the same method name: `calculate_cost(nights)`.

- Abstraction -> Person and Room are abstract base classes. They define abstract methods that child classes must implement.

- Composition -> A Reservation contains a Guest object, a Room object, and HotelService objects. The Hotel class contains collections of guests, rooms, reservations, and services.

- Class Methods -> `Guest.number_of_guests()` is a class method.

- Static Methods -> `Guest.validate_phone()` and `Room.validate_nights()` are static methods.

- Special Methods -> Special methods such as `__str__()`, `__repr__()`, `__len__()`, `__eq__()`, and `__hash__()` are implemented.

- Exception Handling -> Custom exceptions such as `RoomUnavailableError`, `GuestNotFoundError`, `InvalidNightsError`, and `InvalidReservationStatusError` are used.

## EXCEPTIONS

````python
"""
Custom exceptions for the Hotel Reservation and Management System.

This module demonstrates exception handling and custom exception design.
"""


class HotelError(Exception): # Inherits from the built-in Exception class.
    """Base exception for all hotel-related errors."""
    pass

# all custom exceptions inherit from HotelError, which allows for a unified exception handling approach in the hotel management system. Each specific error class provides a clear and descriptive name, making it easier to identify and handle different error scenarios in the application.
class ValidationError(HotelError):
    """Raised when input data is invalid."""
    pass


class GuestNotFoundError(HotelError):
    """Raised when a guest cannot be found."""
    pass


class DuplicateGuestError(HotelError):
    """Raised when a guest already exists in the system."""
    pass


class RoomNotFoundError(HotelError):
    """Raised when a room cannot be found."""
    pass


class RoomUnavailableError(HotelError):
    """Raised when a room cannot be reserved."""
    pass


class InvalidNightsError(HotelError):
    """Raised when the number of nights is invalid."""
    pass


class DuplicateReservationError(HotelError):
    """Raised when a duplicate reservation is attempted."""
    pass


class ReservationNotFoundError(HotelError):
    """Raised when a reservation cannot be found."""
    pass


class ServiceNotFoundError(HotelError):
    """Raised when a hotel service cannot be found."""
    pass


class InvalidReservationStatusError(HotelError):
    """Raised when a reservation operation is not allowed due to its status."""
    pass


class PaymentError(HotelError):
    """Raised when a payment operation fails."""
    pass
````

HotelError(Exception) is the Base Class. It inherits from Python's built-in Exception class. Exception is the base class for all built-in, non-system-exiting exceptions in Python, serving as the standard parent class for most error types like ValueError, TypeError, and KeyError.By making every other error inherit from HotelError, you create a single "umbrella" category.

Enables one to write a single except HotelError: block to catch any error related to the hotel system, or you can catch specific ones like except RoomUnavailableError: if you need to handle that specific scenario differently.
All the other classes inherit from HotelError.

`pass` is used here because the classes don't need to add any new attributes or methods. They simply exist to give a name to a specific type of error.

Data Validation: ValidationError, InvalidNightsError (e.g., trying to book -2 nights).

Guest Management: GuestNotFoundError, DuplicateGuestError (e.g., trying to register a user who already exists).

Room Management: RoomNotFoundError, RoomUnavailableError (e.g., the room is already booked).

Reservations: DuplicateReservationError, ReservationNotFoundError, InvalidReservationStatusError (e.g., trying to check-in a reservation that was already cancelled).

Operations: ServiceNotFoundError, PaymentError.

##  PERSON, GUEST, & STAFF

````python
"""
Person, Guest, and Staff classes.

This module demonstrates:
- Abstraction: Person class inherits from Abstract Base Class.
               Contains @abstractmethod get_role() which forces all child classes to implement this method
               You cannot instantiate Person() directly — you must use Guest or Staff
- Inheritance: Guest and Staff inherit from Person, reusing its attributes and methods.
- Encapsulation: Using @property and @setter decorators to control access to attributes.
                 _name and _id_number are private attributes, accessed through public properties.
                 The @property decorator creates a getter — accessed like person.name.
                 The @name.setter creates a setter with validation logic. Data cannot be set to an invalid value 
                 
- Class attributes
- Class methods
- Static methods
- Special methods: __str__() — User-friendly string representation (called by print())
                   __repr__() — Developer-friendly representation (called in debugging/console)
                   self.__class__.__name__ dynamically gets the class name, so it works for both Guest and Staff
"""


class Person(ABC):
    """
    Abstract base class for all people in the hotel system.

    Demonstrates abstraction through the abstract method get_role().
    """

    def __init__(self, name, id_number):
        self.name = name
        self.id_number = id_number

    @property # defines a getter method that can be accessed like an attribute and return the private value.
    def name(self):
        return self._name

    @name.setter # allows you to define the "set" side of a property for validation and assignment.
    def name(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Name must be a non-empty string.")
        self._name = value.strip()

    @property # defines a getter method that can be accessed like an attribute.
    def id_number(self):
        return self._id_number

    @id_number.setter # allows you to define the "set" side of a property.
    def id_number(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("ID number must be a non-empty string.")
        self._id_number = value.strip()

    @abstractmethod # marks a method as a required blueprint. It forces any child class to implement this method
    def get_role(self):
        """Return the role of the person in the hotel system."""
        pass

    def get_details(self):
        """Return basic person details."""
        return f"{self.get_role()}: {self.name} (ID: {self.id_number})"

    def __str__(self):
        return self.get_details()

    def __repr__(self):
        return f"<{self.__class__.__name__}(name='{self.name}', id_number='{self.id_number}')>"


class Guest(Person):
    """
    Represents a hotel guest.

    Inherits all attributes and methods from Person.
    Uses super().__init__(name, id_number) to call the parent constructor first.
    """

    guest_count = 0 # Shared across all Guest instances. Increments every time a new guest is created: Guest.guest_count += 1

    def __init__(self, name, id_number, phone, guest_type="Regular"):
        super().__init__(name, id_number)
        self.phone = Guest.validate_phone(phone)
        self.guest_type = guest_type # The type of the guest (e.g., Regular, VIP).
        self.registered_on = datetime.now().date() # The date when the guest was registered in the system.
        self._loyalty_points = 0 # Private attribute to track loyalty points. It is read-only from outside the class. Use add_loyalty_points() to modify.
        Guest.guest_count += 1 # Increment the class-level guest count whenever a new Guest instance is created.

    @property
    def phone(self):
        return self._phone

    @phone.setter
    def phone(self, value):
        self._phone = Guest.validate_phone(value)

    @property
    def guest_type(self):
        return self._guest_type

    @guest_type.setter
    def guest_type(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Guest type must be a non-empty string.")
        self._guest_type = value.strip()

    @property # Only a getter, NO setter. Read-only from outside. Must use add_loyalty_points() method to modify.
    def loyalty_points(self):
        return self._loyalty_points

    def add_loyalty_points(self, points):
        """Add loyalty points to the guest account."""
        if not isinstance(points, int) or points < 0:
            raise ValueError("Loyalty points must be zero or a positive integer.")
        self._loyalty_points += points

    def get_role(self): # Method Overriding -> Implements the abstract method from Person with Guest-specific behavior.
        return "Guest"

    @classmethod # It binds a method to the class rather than the instance. Can access class attributes and other class methods. It can be called on the class itself or on an instance of the class i.e Guest.number_of_guests().
    def number_of_guests(cls):
        """Return the total number of registered guests."""
        return cls.guest_count

    @staticmethod #  It defines a method that does not receive an implicit first argument. It cannot access or modify class state or instance state. It is used for utility functions that are related to the class but do not need to access any class or instance-specific data i.e Guest.validate_phone("+254700111222").
    def validate_phone(phone):
        """Validate phone number format."""
        if not isinstance(phone, str):
            raise ValueError("Phone number must be a string.")

        phone = phone.strip()

        if len(phone) < 7:
            raise ValueError("Phone number is too short.")

        return phone

    def __str__(self):
        return (
            f"Guest: {self.name} | ID: {self.id_number} | "
            f"Phone: {self.phone} | Type: {self.guest_type}"
        )


class Staff(Person):
    """
    Represents hotel staff.

    Inherits all attributes and methods from Person.
    Adds a role attribute specific to staff.
    Overrides get_role() to return f"Staff - {self.role}"
    Overrides __str__() for staff-specific display
    """

    def __init__(self, name, id_number, role):
        super().__init__(name, id_number)
        self.role = role

    @property
    def role(self):
        return self._role

    @role.setter
    def role(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Staff role must be a non-empty string.")
        self._role = value.strip()

    def get_role(self): # Method Overriding -> Implements the abstract method from Person with Staff-specific behavior.
        return f"Staff - {self.role}"

    def __str__(self):
        return f"Staff: {self.name} | ID: {self.id_number} | Role: {self.role}"
````

## ROOMS

````python
"""
Room classes.

This module demonstrates:
- Abstraction
- Inheritance
- Polymorphism
- Method overriding
- Encapsulation
- Static methods
- Special methods
"""


class Room(ABC):
    """
    Abstract base class for all hotel rooms.

    Child classes must implement:
    - get_description()
    - calculate_cost(nights)
    """
    # Class states that transitions when a new event occurs. These are shared across all Room instances and represent the current state of a room.
    AVAILABLE = "AVAILABLE"
    RESERVED = "RESERVED"
    OCCUPIED = "OCCUPIED"
    CLEANING = "CLEANING"
    MAINTENANCE = "MAINTENANCE"
    
    # class attributes shared across all Room instances.

    def __init__(self, room_number, base_price):
        self.room_number = room_number # The room number is a unique identifier for the room. It is validated to ensure it is a non-empty string.
        self.base_price = base_price # The base price is the cost per night for the room. It is validated to ensure it is a positive number.
        self._status = Room.AVAILABLE # The initial status of a room is set to AVAILABLE when it is created. This private attribute is read-only from outside the class and can only be changed through specific methods like make_reserved(), make_occupied(), make_cleaning(), and make_available().

    @property
    def room_number(self):
        return self._room_number

    @room_number.setter
    def room_number(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Room number must be a non-empty string.")
        self._room_number = value.strip()

    @property
    def base_price(self):
        return self._base_price

    @base_price.setter
    def base_price(self, value):
        if not isinstance(value, (int, float)) or value <= 0:
            raise ValueError("Base price must be greater than zero.")
        self._base_price = float(value)

    @property # status is read-only (getter only, no setter) therefore can only change through specific methods: make_reserved(), make_occupied(), make_cleaning(), make_available().
    def status(self):
        return self._status

    def is_available(self):
        """Check whether the room is available for booking."""
        return self._status == Room.AVAILABLE

    def make_reserved(self):
        """Reserve the room, otherwise raise RoomUnavailableError."""
        if not self.is_available():
            raise RoomUnavailableError(
                f"Room {self.room_number} is not available for reservation."
            )
        self._status = Room.RESERVED 

    def make_occupied(self):
        """Mark the room as occupied, otherwise raise RoomUnavailableError."""
        if self._status != Room.RESERVED:
            raise RoomUnavailableError(
                f"Room {self.room_number} must be reserved before check-in."
            )
        self._status = Room.OCCUPIED

    def make_cleaning(self):
        """Mark the room as cleaning."""
        self._status = Room.CLEANING

    def make_available(self):
        """Mark the room as available."""
        self._status = Room.AVAILABLE

    @staticmethod # It defines a method that does not receive an implicit first argument. It cannot access or modify class state or instance state. It is used for utility functions that are related to the class but do not need to access any class or instance-specific data i.e Room.validate_nights(3).
    def validate_nights(nights):
        """Validate number of nights."""
        if isinstance(nights, bool) or not isinstance(nights, int):  # checks if an object is an instance of a particular class or data type.
            raise InvalidNightsError("Number of nights must be an integer.")

        if nights <= 0 or nights > 365:
            raise InvalidNightsError(
                "Number of nights must be a positive integer not greater than 365."
            )

        return nights

    @abstractmethod
    def get_description(self):
        """Return a description of the room."""
        pass

    @abstractmethod
    def calculate_cost(self, nights):
        """Calculate the accommodation cost for the room."""
        pass

    def __str__(self): # User-friendly string representation of the room, showing its number, status, and base price.
        return (
            f"Room {self.room_number} | Status: {self.status} | "
            f"Base Price: {self.base_price:.2f}"
        )

    def __repr__(self): # Developer-friendly representation of the room, showing its class name, number, and status. Useful for debugging and logging.
        return (
            f"<{self.__class__.__name__}(room_number='{self.room_number}', "
            f"status='{self.status}')>"
        )


class StandardRoom(Room):
    """
    Represents a standard hotel room.
    """

    def __init__(self, room_number, base_price, has_tv=True):
        super().__init__(room_number, base_price)
        self.has_tv = has_tv

    def get_description(self): # Method Overriding -> Implements the abstract method from Room with StandardRoom-specific behavior.
        tv_info = "with TV" if self.has_tv else "without TV"
        return f"Standard Room {self.room_number} {tv_info}"

    def calculate_cost(self, nights): # Method Overriding -> Implements the abstract method from Room with StandardRoom-specific behavior.
        """
        Implements polymorphism. The same method name calculate_cost() behaves 
        differently depending on the object type. 
        You can call room.calculate_cost(3) on any room without knowing its specific type.
        
        """
        Room.validate_nights(nights)
        return round(self.base_price * nights, 2)


class DeluxeRoom(Room):
    """
    Represents a deluxe hotel room.
    """

    def __init__(self, room_number, base_price, includes_breakfast=True):
        super().__init__(room_number, base_price)
        self.includes_breakfast = includes_breakfast

    def get_description(self): # Method Overriding -> Implements the abstract method from Room with DeluxeRoom-specific behavior.
        breakfast_info = "with breakfast" if self.includes_breakfast else "without breakfast"
        return f"Deluxe Room {self.room_number} {breakfast_info}"

    def calculate_cost(self, nights): # Method Overriding -> Implements the abstract method from Room with DeluxeRoom-specific behavior.
        Room.validate_nights(nights)
        return round(self.base_price * nights * 1.15, 2)


class Suite(Room):
    """
    Represents a luxury hotel suite.
    """

    def __init__(self, room_number, base_price, lounge_access=True):
        super().__init__(room_number, base_price)
        self.lounge_access = lounge_access

    def get_description(self): # Method Overriding -> Implements the abstract method from Room with Suite-specific behavior.
        lounge_info = "with lounge access" if self.lounge_access else "without lounge access"
        return f"Executive Suite {self.room_number} {lounge_info}"

    def calculate_cost(self, nights): # Method Overriding -> Implements the abstract method from Room with Suite-specific behavior.
        Room.validate_nights(nights)

        cost = self.base_price * nights * 1.25

        if self.lounge_access:
            cost += 500 * nights

        return round(cost, 2)
````

## HOTEL SERVICES

````python
"""
HotelService class.

This module demonstrates:
- Encapsulation
- Class attributes
- Class methods
- Special methods
"""


class HotelService:
    """
    Represents an additional hotel service such as laundry,
    breakfast, airport pickup, or room service.
    """

    _next_service_id = 100 # Class attribute to generate unique service IDs. It tracks the next available ID.

    def __init__(self, name, price):
        self.service_id = HotelService._generate_service_id()
        self.name = name
        self.price = price

    @classmethod
    def _generate_service_id(cls): 
        """
        Class method to generate a unique service ID. 
        It increments the class attribute _next_service_id for each new service.
        Every new service gets a unique ID automatically (SRV-101, SRV-102).
        """
        cls._next_service_id += 1
        return f"SRV-{cls._next_service_id}"

    # Both name and price have property getters/setters with validation.
    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Service name must be a non-empty string.")
        self._name = value.strip()

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if not isinstance(value, (int, float)) or value < 0:
            raise ValueError("Service price must be zero or greater.")
        self._price = float(value)

    def __str__(self): # User-friendly string representation of the service, showing its ID, name, and price.
        return f"{self.service_id}: {self.name} - {self.price:.2f}"

    def __repr__(self): # Developer-friendly representation of the service, showing its class name, ID, and name. Useful for debugging and logging.
        return f"<HotelService(service_id='{self.service_id}', name='{self.name}')>"

    def __eq__(self, other): # —> Defines equality (two services are equal if they have the same ID and share the same memory address). This allows you to compare two HotelService instances using the == operator.
        if not isinstance(other, HotelService):
            return NotImplemented
        return self.service_id == other.service_id

    def __hash__(self): # —> Makes objects usable in sets and as dictionary keys. 
        return hash(self.service_id)
    
    # __hash__ and __eq__ work together to ensure that services with the same ID are treated as equal and can be used in sets and as dictionary keys.
    # These prevent duplicate services from being added to a reservation
````

## PAYMENT

````python
"""
Payment classes.

This module demonstrates:
- Abstraction
- Inheritance
- Polymorphism
- Encapsulation
- Class attributes
- Class methods
- Static methods
- Special methods
"""


class Payment(ABC):
    """
    Abstract base class for all payment types. Dictates what every payment must have and do.

    Child classes must implement:
    - get_payment_method()
    - process_payment()
    """

    _next_payment_id = 5000 # Class Attribute: Counter for entire payment system.

    # Constants that act as a strict state machine to prevent invalid transitions between payment states. The payment can only transition from PENDING to COMPLETED, FAILED, or REFUNDED. Once a payment is marked as COMPLETED, it cannot be changed back to PENDING or FAILED. Similarly, a REFUNDED payment cannot be changed back to any other state.
    PENDING = "PENDING"
    COMPLETED = "COMPLETED"
    FAILED = "FAILED"
    REFUNDED = "REFUNDED"

    def __init__(self, amount):
        self._payment_id = Payment._generate_payment_id() # instance attribute unique to each payment. It is generated using the class method _generate_payment_id() which increments the class attribute _next_payment_id for each new payment.
        self.amount = amount
        self._payment_date = datetime.now() # encapsulation, protected attribute: The payment date is set to the current date and time when the payment is created. It is a read-only property, so it cannot be modified after creation.
        self._status = Payment.PENDING # Default status is PENDING until the payment is processed. It can change to COMPLETED, FAILED, or REFUNDED based on the payment outcome.

    @classmethod
    def _generate_payment_id(cls):
        """
        Class method that modifies the class attribute _next_payment_id.
        class attribute _next_payment_id is used to generate unique payment IDs. 
        It is incremented each time a new payment is created. 
        The @classmethod _generate_payment_id() generates the next payment ID in the format PAY-5001, PAY-5002, etc.
        """
        cls._next_payment_id += 1 # class attribute 
        return f"PAY-{cls._next_payment_id}"

    #encapsulation: Using @property and @setter decorators to control access to attributes.
    @property
    def payment_id(self):
        return self._payment_id

    @property # read only property. No setter.  ensuring nobody can maliciously change a payment ID or mark a failed payment as completed without going through the proper methods.
    def payment_date(self):
        return self._payment_date

    @property
    def status(self):
        return self._status

    @property
    def amount(self):
        return self._amount

    @amount.setter #It includes validation logic to ensure that the amount is a positive number.
    def amount(self, value):
        if isinstance(value, bool) or not isinstance(value, (int, float)):
            raise PaymentError("Payment amount must be a number.")

        if value <= 0:
            raise PaymentError("Payment amount must be greater than zero.")

        self._amount = float(value)

    @abstractmethod # Forces any child class to implement this method
    def get_payment_method(self):
        """Return the payment method name."""
        pass

    @abstractmethod
    def process_payment(self):
        """Process the payment."""
        pass

    def _base_receipt_lines(self):
        return [
            "=" * 45,
            "PAYMENT RECEIPT",
            "=" * 45,
            f"Payment ID: {self.payment_id}",
            f"Payment Method: {self.get_payment_method()}",
            f"Amount Paid: {self.amount:.2f}",
            f"Payment Status: {self.status}",
            f"Payment Date: {self.payment_date.strftime('%Y-%m-%d %H:%M:%S')}"
        ]

    def _additional_receipt_lines(self):
        """
        Defines the structure of the receipt
        Child classes can override this method to add extra receipt details.
        """
        return []

    def get_receipt(self):
        """Return a formatted payment receipt."""
        lines = self._base_receipt_lines()
        lines.extend(self._additional_receipt_lines())
        lines.append("=" * 45)
        return "\n".join(lines)

    def __str__(self):
        return (
            f"{self.payment_id} | {self.get_payment_method()} | "
            f"Amount: {self.amount:.2f} | Status: {self.status}"
        )

    def __repr__(self):
        return (
            f"<{self.__class__.__name__}(payment_id='{self.payment_id}', "
            f"amount={self.amount}, status='{self.status}')>"
        )


class CashPayment(Payment):
    """
    Represents cash payment.
    """

    def __init__(self, amount, cash_received=None):
        super().__init__(amount) # inherits the amount from the parent Payment class and initializes the cash_received attribute. If cash_received is not provided, it defaults to the amount.
        self.cash_received = cash_received if cash_received is not None else amount

    @property
    def cash_received(self): # tracks cash_received 
        return self._cash_received

    @cash_received.setter # 
    def cash_received(self, value):
        if isinstance(value, bool) or not isinstance(value, (int, float)):
            raise PaymentError("Cash received must be a number.")

        if value < 0:
            raise PaymentError("Cash received cannot be negative.")

        self._cash_received = float(value)

    @property
    def change_due(self): #  calculates change_due using a getter property
        if self.cash_received >= self.amount:
            return round(self.cash_received - self.amount, 2)
        return 0.00

    def get_payment_method(self):
        return "Cash"

    def process_payment(self): # checks if cash_received < amount. If so, it changes status to FAILED and raises an error. Otherwise, it marks it COMPLETED.
        if self.cash_received < self.amount:
            self._status = Payment.FAILED
            raise PaymentError(
                "Insufficient cash received to complete this payment."
            )

        self._status = Payment.COMPLETED
        return self.get_receipt()

    def _additional_receipt_lines(self): # to print the cash received and change given.
        return [
            f"Cash Received: {self.cash_received:.2f}",
            f"Change Due: {self.change_due:.2f}"
        ]


class CardPayment(Payment):
    """
    Represents card payment.
    """

    def __init__(self, amount, card_number, expiry_date, cvv):
        super().__init__(amount)
        self._card_number = CardPayment.validate_card_number(card_number)
        self._expiry_date = CardPayment.validate_expiry_date(expiry_date)
        self._cvv = CardPayment.validate_cvv(cvv)

    @property # Sets the card number as a private attribute and provides a masked version for display. The actual card number is never exposed in full for security reasons.
    def masked_card_number(self):
        last_four_digits = self._card_number[-4:]
        return f"****-****-****-{last_four_digits}"

    @staticmethod
    def validate_card_number(card_number):
        """ Helper method to validate card number format of a payment card number, raising a custom PaymentError 
        if the input fails any structural checks and returning the cleaned numeric string if 
        successful."""
        if not isinstance(card_number, str):
            raise PaymentError("Card number must be a string.")

        cleaned_card_number = card_number.replace(" ", "").replace("-", "")

        if not cleaned_card_number.isdigit():
            raise PaymentError("Card number must contain only digits.")

        if len(cleaned_card_number) < 12 or len(cleaned_card_number) > 19:
            raise PaymentError("Card number must be between 12 and 19 digits.")

        # Inline Luhn Algorithm validation
        total = 0
        for i, char in enumerate(cleaned_card_number[::-1]): # Reversal and enumeration of the cleaned card number to apply the Luhn algorithm. The algorithm doubles every second digit from the right, subtracts 9 if the result is greater than 9, and sums all digits. If the total modulo 10 is not zero, the card number is invalid.
            digit = int(char)
            if i % 2 == 1:
                digit *= 2
                if digit > 9:
                    digit -= 9
            total += digit

        if total % 10 != 0:
            raise PaymentError("Invalid card number (checksum failed).")
        
        return cleaned_card_number

    @staticmethod
    def validate_expiry_date(expiry_date):
        """Validate card expiry date in MM/YY format."""
        if not isinstance(expiry_date, str):
            raise PaymentError("Expiry date must be a string.")

        expiry_date = expiry_date.strip() # Strips outer whitespace and splits the string on /.
        parts = expiry_date.split("/")

        if len(parts) != 2:
            raise PaymentError("Expiry date must be in MM/YY format.")

        month_string = parts[0].strip()
        year_string = parts[1].strip()

        if len(month_string) != 2 or len(year_string) != 2: # Confirms there are exactly two parts (month and year)
            raise PaymentError("Expiry date must be in MM/YY format.")

        try: # Converts both components to integers.
            month = int(month_string)
            year = int(year_string)
        except ValueError:
            raise PaymentError("Expiry date must contain valid numbers.")

        if month < 1 or month > 12: # Restricts month to the valid range ($1$ to $12$).
            raise PaymentError("Expiry month must be between 01 and 12.")

        now = datetime.now()
        current_year = now.year % 100
        current_month = now.month

        # Rejects any date prior to the current month of the current year
        if year < current_year or (year == current_year and month < current_month):
            raise PaymentError("Card is expired.")

        return expiry_date

    @staticmethod
    def validate_cvv(cvv):
        """Validate card CVV - (Card Verification Value)."""
        if not isinstance(cvv, str): # Ensures input is a string and removes leading/trailing spaces.
            raise PaymentError("CVV must be a string.")

        cvv = cvv.strip()

        if not cvv.isdigit():
            raise PaymentError("CVV must contain only digits.")

        if len(cvv) < 3 or len(cvv) > 4:
            raise PaymentError("CVV must be 3 or 4 digits.")

        return cvv

    def get_payment_method(self): # Method Overriding -> Implements the abstract method from Payment with CardPayment-specific behavior. It returns a string indicating the payment method.
        return "Card"

    def process_payment(self):

        """
        Simulates card authorization check.
        """
        if not self._card_number or not self._cvv:
            self._status = Payment.FAILED
            raise PaymentError("Card authorization failed.")

        self._status = Payment.COMPLETED
        return self.get_receipt()

    def _additional_receipt_lines(self):
        return [
            f"Card Number: {self.masked_card_number}",
            f"Expiry Date: {self._expiry_date}"
        ]


class MobilePayment(Payment):
    """
    Represents mobile money payment such as M-Pesa, Airtel Money or other mobile wallet providers.
    """

    def __init__(self, amount, phone_number, provider):
        super().__init__(amount) # Inheritance: Calls the parent Payment class constructor to initialize the amount and other attributes.
        self.phone_number = phone_number
        self.provider = provider

    @property
    def phone_number(self):
        return self._phone_number

    @phone_number.setter
    def phone_number(self, value):
        if not isinstance(value, str):
            raise PaymentError("Phone number must be a string.")

        value = value.strip()

        if len(value) < 10:
            raise PaymentError("Phone number is too short.")

        self._phone_number = value

    @property
    def provider(self):
        return self._provider

    @provider.setter
    def provider(self, value):
        if not isinstance(value, str) or not value.strip():
            raise PaymentError("Mobile payment provider must be a non-empty string.")

        self._provider = value.strip()
        
    # Polymorphism: The same method name get_payment_method() behaves differently depending on the object type. You can call payment.get_payment_method() on any payment without knowing its specific type.
    def get_payment_method(self): # method overriding to provide a specific implementation for mobile payments. It returns a string indicating the payment method and the provider.
        return f"Mobile - {self.provider}"

    def process_payment(self): 
        """
        Simulate mobile money STK push or wallet authorization.
        """
        if len(self.phone_number) < 10 or not self.provider:
            self._status = Payment.FAILED
            raise PaymentError("Mobile payment authorization failed.")

        self._status = Payment.COMPLETED
        return self.get_receipt()

    def _additional_receipt_lines(self): # implements the abstract method from Payment with MobilePayment-specific behavior. It returns a list of strings that will be appended to the receipt, providing details about the mobile payment provider and the phone number used for the transaction.
        return [
            f"Provider: {self.provider}",
            f"Phone Number: {self.phone_number}"
        ]
````

## RESERVATION

````python
"""
Reservation class.

This module demonstrates:
- Composition - Reservation contains Guest, Room, list of HotelService, and list of Payment objects.
- Encapsulation
- Exception handling
- Special methods
- Class methods
"""


class Reservation:
    """
    Represents a reservation made by a guest for a room.

    Demonstrates composition because a Reservation contains:
    - Guest object
    - Room object
    - HotelService objects
    - Payment objects
    """

    _next_reservation_id = 1000

    PENDING = "PENDING"
    ACTIVE = "ACTIVE"
    CANCELLED = "CANCELLED"
    COMPLETED = "COMPLETED"

    UNPAID = "UNPAID"
    PARTIALLY_PAID = "PARTIALLY_PAID"
    PAID = "PAID"

    def __init__(self, guest, room, nights, check_in_date=None):
        if not isinstance(guest, Guest):
            raise TypeError("guest must be an instance of Guest.")

        if not isinstance(room, Room):
            raise TypeError("room must be an instance of Room.")

        Room.validate_nights(nights)

        if not room.is_available():
            raise RoomUnavailableError(
                f"Room {room.room_number} is not available for reservation."
            )

        self._reservation_id = Reservation._generate_reservation_id() # generates a unique reservation ID using a class method that increments a class attribute. This ensures that each reservation has a distinct identifier.
        self._guest = guest #protected attributes are used to encapsulate the guest and room objects, preventing direct modification from outside the class. This promotes data integrity and allows controlled access through properties.
        self._room = room
        self.nights = nights
        self._check_in_date = check_in_date if check_in_date else datetime.now().date() # If no check-in date is provided, it defaults to the current date. This allows for flexibility in reservation creation while ensuring that a valid date is always set.
        self._status = Reservation.PENDING # The initial status of the reservation is set to PENDING.
        self._services = [] # Initializes an empty list attribute to store future add-on service objects
        self._payments = [] # Initializes an empty list attribute to track all payment objects tied to this reservation.
        self._payment_status = Reservation.UNPAID # Sets the initial financial status to UNPAID using a class-level constant.
        self._created_on = datetime.now() # Captures and records the exact date and time the reservation was created.

        self._room.make_reserved()# Marks the associated room as RESERVED, preventing other guests from booking it until the reservation is either completed or cancelled.

    @classmethod
    def _generate_reservation_id(cls): # internal helper method used to generate sequential reservation identifiers. 
        cls._next_reservation_id += 1
        return f"RES-{cls._next_reservation_id}" # The reservation ID is formatted as a string with a prefix "RES-" followed by the incremented number, ensuring uniqueness across all reservations.

    # Encapsulated read only properties for reservation attributes. These properties provide controlled access to the internal state of the reservation object, allowing external code to retrieve information without directly modifying the underlying data.
    @property
    def reservation_id(self):
        return self._reservation_id

    @property
    def guest(self):
        return self._guest

    @property
    def room(self):
        return self._room

    @property
    def check_in_date(self):
        return self._check_in_date

    @property
    def status(self):
        return self._status

    @property
    def created_on(self):
        return self._created_on

    @property
    def services(self):
        """Return a tuple of services to prevent external modification via append()."""
        return tuple(self._services)

    @property
    def payments(self):
        """Return a tuple of payments to prevent external modification via append()."""
        return tuple(self._payments)

    @property
    def payment_status(self): # Exposes current financial status
        return self._payment_status

    # Encapsulated property for nights with validation. The setter ensures that the number of nights is always a valid positive integer, maintaining the integrity of the reservation's duration.
    @property
    def nights(self):
        return self._nights

    @nights.setter
    def nights(self, value):
        Room.validate_nights(value)
        self._nights = value

    @property
    def amount_paid(self): # Calculates the sum of all successful payments logged against the reservation.
        total_paid = sum(
            payment.amount for payment in self._payments
            if payment.status == Payment.COMPLETED
        ) # Filters self._payments for payments where status == Payment.COMPLETED and sums their amounts.
        return round(total_paid, 2)

    @property
    def balance_due(self): # Calculates the outstanding unpaid balance.
        balance = self.calculate_total_charge() - self.amount_paid
        return round(max(balance, 0), 2)

    @property # helper to check if the reservation is fully paid.
    def is_paid(self):
        return self.balance_due <= 0 # Returns True if no balance remains unpaid; otherwise returns False.

    def add_service(self, service):
        """Add an additional hotel service to the reservation i.e laundry, breakfast."""
        if not isinstance(service, HotelService):
            raise TypeError("service must be an instance of HotelService.")

        if self._status in [Reservation.CANCELLED, Reservation.COMPLETED]: # Checks if the reservation is in a state that allows adding services. If the reservation is either CANCELLED or COMPLETED, it raises an InvalidReservationStatusError to prevent modifications to finalized reservations.
            raise InvalidReservationStatusError(
                "Services cannot be added to a cancelled or completed reservation."
            )

        if service in self._services: # Checks if the service is already in the list of services for this reservation. If it is, it raises a ValueError to prevent duplicate services from being added.
            raise ValueError(
                f"Service {service.name} has already been added to this reservation."
            )

        self._services.append(service)

    def record_payment(self, payment):
        """Record a completed payment against this reservation bill."""
        if not isinstance(payment, Payment): # Ensures input is an instance of Payment or its subclasses.
            raise TypeError("payment must be an instance of Payment.")

        if self._status == Reservation.CANCELLED: #Checks if reservation is cancelled
            raise InvalidReservationStatusError(
                "Payments cannot be recorded for a cancelled reservation."
            )

        if payment.status != Payment.COMPLETED: # Verifies that the payment has already been authorized/completed.
            raise PaymentError("Only completed payments can be recorded.")

        projected_total_paid = round(self.amount_paid + payment.amount, 2) #Calculates what total paid amount would be if this payment is added.
        total_charge = round(self.calculate_total_charge(), 2) # Fetches current total bill amount.

        if projected_total_paid > total_charge + 0.01: # Checks if payment exceeds total bill (allowing a 1-cent rounding buffer).
            raise PaymentError("Payment amount exceeds outstanding balance.") # Blocks payments that overpay the reservation.

        self._payments.append(payment) # Appends valid payment object to internal _payments list.
        self._update_payment_status() # Triggers helper method to update internal financial status flag.

    def _update_payment_status(self): #Helper method that re-evaluates _payment_status. 
        """Update payment status based on amount paid."""
        total_charge = self.calculate_total_charge() # Retrieves current total charge.

        if self.amount_paid <= 0:
            self._payment_status = Reservation.UNPAID # Sets payment status to UNPAID if no payments have been made.
        elif self.amount_paid >= total_charge:
            self._payment_status = Reservation.PAID # Sets payment status to PAID if full amount has been settled.
        else:
            self._payment_status = Reservation.PARTIALLY_PAID # Sets payment status to PARTIALLY_PAID if partial amount has been settled.

    def calculate_room_charge(self):
        """Calculate accommodation charge only."""
        return self._room.calculate_cost(self.nights)

    def calculate_service_charge(self):
        """Calculate total service charge."""
        total_service_charge = sum(service.price for service in self._services)
        return round(total_service_charge, 2)

    def calculate_total_charge(self):
        """Calculate total bill amount."""
        total = self.calculate_room_charge() + self.calculate_service_charge()
        return round(total, 2)

    def cancel(self):
        """Cancel the reservation and releases reserved room back to inventory."""
        if self._status == Reservation.CANCELLED: #Prevents re-cancelling an already cancelled reservation.
            raise InvalidReservationStatusError(
                f"Reservation {self.reservation_id} is already cancelled."
            )

        if self._status == Reservation.COMPLETED: # Prevents cancelling a stay that has already been finished.
            raise InvalidReservationStatusError(
                f"Reservation {self.reservation_id} is completed and cannot be cancelled."
            )

        self._status = Reservation.CANCELLED
        self._room.make_available()
        """Releases the reserved room back to the inventory."""

    def check_in(self):
        """Check the guest into the reservation."""
        if self._status == Reservation.CANCELLED:
            raise InvalidReservationStatusError(
                f"Reservation {self.reservation_id} is cancelled and cannot check in."
            )

        if self._status == Reservation.COMPLETED: # Prevents checking in a reservation that has already been completed.
            raise InvalidReservationStatusError(
                f"Reservation {self.reservation_id} is already completed."
            )

        self._room.make_occupied() # Updates reservation state to ACTIVE.
        self._status = Reservation.ACTIVE # Sets reservation status to ACTIVE.

    def check_out(self):
        """Check the guest out of the reservation."""
        if self._status != Reservation.ACTIVE: # Ensures check-out only happens for currently active check-ins.
            raise InvalidReservationStatusError(
                f"Reservation {self.reservation_id} must be active before check-out."
            )

        self._room.make_available() # Releases room back to available inventory for future bookings.
        self._status = Reservation.COMPLETED #Marks reservation status as COMPLETED.

    def get_bill_summary(self):
        """Generate a formatted final bill summary."""
        lines = []

        lines.append("=" * 45)
        lines.append(f"FINAL BILL - {self.reservation_id}")
        lines.append("=" * 45)
        lines.append(f"Guest: {self.guest.name}")
        lines.append(f"Guest ID: {self.guest.id_number}")
        lines.append(f"Room: {self.room.room_number}")
        lines.append(f"Room Type: {self.room.get_description()}")
        lines.append(f"Nights: {self.nights}")
        lines.append(f"Room Charge: {self.calculate_room_charge():.2f}")

        if self._services:
            lines.append("Services:")
            for service in self._services:
                lines.append(f"  - {service.name}: {service.price:.2f}")
        else:
            lines.append("Services: None")

        lines.append(f"Service Charge: {self.calculate_service_charge():.2f}")
        lines.append("-" * 45)
        lines.append(f"Total Charge: {self.calculate_total_charge():.2f}")
        lines.append(f"Payment Status: {self.payment_status}")
        lines.append(f"Amount Paid: {self.amount_paid:.2f}")
        lines.append(f"Balance Due: {self.balance_due:.2f}")
        lines.append("=" * 45)

        return "\n".join(lines)

    def __len__(self):
        """Return the number of nights in the reservation."""
        return self.nights

    def __str__(self):
        return (
            f"{self.reservation_id} | {self.guest.name} | "
            f"Room {self.room.room_number} | {self.nights} nights | "
            f"{self.status} | Payment: {self.payment_status} | "
            f"Total: {self.calculate_total_charge():.2f}"
        )

    def __repr__(self):
        return (
            f"<Reservation(reservation_id='{self.reservation_id}', "
            f"guest='{self.guest.name}', room='{self.room.room_number}')>"
        )
````

## HOTEL CONTROLLER

````python
"""
Hotel controller class.

This module demonstrates:
- Composition -> has-a collection of guests, staff, rooms, reservations, and services stored in private lists (_guests, _rooms, etc.). 
- Encapsulation -> Private attributes and methods are used to control access to the internal state of the hotel.
- Exception handling
- Controller class design -> The Hotel class acts as a central controller for managing the hotel system, providing methods to register guests and staff, add rooms and services, and retrieve entities by their identifiers. It encapsulates the logic for managing the hotel's operations and ensures data integrity through validation and controlled access.
"""


class Hotel:
    """
    Central controller class for the hotel system.

    Manages:
    - Guests
    - Staff
    - Rooms
    - Reservations
    - Hotel services
    - Payments
    """

    def __init__(self, name):
        self.name = name
        self._guests = []
        self._staff = []
        self._rooms = []
        self._reservations = []
        self._services = []

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, value):
        if not isinstance(value, str) or not value.strip():
            raise ValueError("Hotel name must be a non-empty string.")
        self._name = value.strip()

    @property
    def guests(self):
        return tuple(self._guests) # Returns a tuple of guests to prevent external modification via append() or remove(). This encapsulation ensures that the internal list of guests cannot be altered directly from outside the class, maintaining data integrity.

    @property
    def staff(self):
        return tuple(self._staff)

    @property
    def rooms(self):
        return tuple(self._rooms)

    @property
    def reservations(self):
        return tuple(self._reservations)

    @property
    def services(self):
        return tuple(self._services)

    def register_guest(self, name, id_number, phone, guest_type="Regular"):
        """Register a new guest."""
        for guest in self._guests:
            if guest.id_number == id_number:
                raise DuplicateGuestError(
                    f"Guest with ID {id_number} already exists."
                ) # Loops through the hotel's private _guests list and compares each existing guest's id_number against the incoming id_number. If a match is found, it raises a DuplicateGuestError to prevent duplicate guest entries.

        guest = Guest(name, id_number, phone, guest_type) # Creates a new Guest instance using the validated inputs (delegating any field-level validation, such as phone format or name checks, to the Guest class constructor)
        self._guests.append(guest) # Adds the newly created Guest object to the hotel's private _guests list, effectively registering the guest in the system.
        return guest

    def register_staff(self, name, id_number, role):
        """Register a new staff member."""
        for staff in self._staff:
            if staff.id_number == id_number:
                raise ValueError(f"Staff with ID {id_number} already exists.")
        # Validates the input for the new staff member.
        if not isinstance(name, str) or not name.strip():
            raise ValueError("Staff name must be a non-empty string.") # Ensures the staff member's name is a non-empty string.

        if not isinstance(id_number, str) or not id_number.strip():
            raise ValueError("Staff ID must be a non-empty string.")

        if not isinstance(role, str) or not role.strip():
            raise ValueError("Staff role must be a non-empty string.")

        staff = Staff(name, id_number, role) # Creates a new Staff instance using the validated inputs (delegating any field-level validation, such as name checks, to the Staff class constructor). 
        self._staff.append(staff)
        return staff

    def add_room(self, room_type, room_number, base_price):
        """Add a room to the hotel."""
        for room in self._rooms:
            if room.room_number == room_number:
                raise ValueError(f"Room {room_number} already exists.")

        if not isinstance(room_type, str) or not room_type.strip():
            raise ValueError("Room type must be a non-empty string.")

        room_type = room_type.strip().lower()

        if room_type == "standard":
            room = StandardRoom(room_number, base_price) # Creates a new StandardRoom instance using the provided room_number and base_price. The StandardRoom class is assumed to be a subclass of Room that implements specific behavior for standard rooms.
        elif room_type == "deluxe":
            room = DeluxeRoom(room_number, base_price) # Creates a new DeluxeRoom instance using the provided room_number and base_price. The DeluxeRoom class is assumed to be a subclass of Room that implements specific behavior for deluxe rooms.
        elif room_type == "suite":
            room = Suite(room_number, base_price)# Creates a new Suite instance using the provided room_number and base_price. The Suite class is assumed to be a subclass of Room that implements specific behavior for suite rooms.
        else:
            raise ValueError(
                "Invalid room type. Use 'standard', 'deluxe', or 'suite'."
            )

        self._rooms.append(room)
        return room

    def add_service(self, name, price):
        """Add a hotel service."""
        service = HotelService(name, price) # Creates a new HotelService instance using the provided name and price. The HotelService class is assumed to encapsulate the details of a service offered by the hotel, such as its name and cost.
        self._services.append(service)
        return service

    def get_guest_by_id(self, id_number):
        """Retrieve a guest by ID."""
        id_number = str(id_number).strip()

        for guest in self._guests:
            if guest.id_number == id_number:
                return guest

        raise GuestNotFoundError(f"Guest with ID {id_number} was not found.")

    def get_room_by_number(self, room_number):
        """Retrieve a room by room number."""
        room_number = str(room_number).strip()

        for room in self._rooms:
            if room.room_number == room_number:
                return room

        raise RoomNotFoundError(f"Room {room_number} was not found.")

    def get_reservation_by_id(self, reservation_id):
        """Retrieve a reservation by reservation ID."""
        reservation_id = str(reservation_id).strip()

        for reservation in self._reservations:
            if reservation.reservation_id == reservation_id:
                return reservation

        raise ReservationNotFoundError(
            f"Reservation {reservation_id} was not found."
        )

    def get_service_by_id(self, service_id):
        """Retrieve a service by service ID."""
        service_id = str(service_id).strip()

        for service in self._services:
            if service.service_id == service_id:
                return service

        raise ServiceNotFoundError(f"Service {service_id} was not found.")

    def search_rooms(self, room_number=None, room_type=None, available_only=True):
        """
        Search rooms by room number, room type, and availability.
        """
        results = self._rooms.copy()

        if room_number:
            room_number = str(room_number).strip()
            results = [
                room for room in results
                if room.room_number == room_number
            ]

        if room_type:
            if not isinstance(room_type, str) or not room_type.strip():
                raise ValueError("Room type must be a non-empty string.")

            room_type = room_type.strip().lower()

            if room_type == "standard":
                results = [
                    room for room in results
                    if isinstance(room, StandardRoom)
                ]
            elif room_type == "deluxe":
                results = [
                    room for room in results
                    if isinstance(room, DeluxeRoom)
                ]
            elif room_type == "suite":
                results = [
                    room for room in results
                    if isinstance(room, Suite)
                ]
            else:
                raise ValueError(
                    "Invalid room type. Use 'standard', 'deluxe', or 'suite'."
                )

        if available_only:
            results = [room for room in results if room.is_available()]

        return results

    def make_reservation(self, guest_id, room_number, nights):
        """Create a reservation."""
        guest = self.get_guest_by_id(guest_id) # Retrieves the Guest object corresponding to the provided guest_id. If no matching guest is found, a GuestNotFoundError is raised.
        room = self.get_room_by_number(room_number) # Retrieves the Room object corresponding to the provided room_number. If no matching room is found, a RoomNotFoundError is raised.

        for reservation in self._reservations:
            if (
                reservation.guest.id_number == guest.id_number
                and reservation.room.room_number == room.room_number
                and reservation.status in [Reservation.PENDING, Reservation.ACTIVE]
            ):
                raise DuplicateReservationError(
                    f"Guest {guest.id_number} already has an active reservation "
                    f"for room {room.room_number}."
                )

        reservation = Reservation(guest, room, nights)
        self._reservations.append(reservation)

        return reservation

    def cancel_reservation(self, reservation_id):
        """Cancel a reservation."""
        reservation = self.get_reservation_by_id(reservation_id)
        reservation.cancel()
        return reservation

    def check_in(self, reservation_id):
        """Check a guest in."""
        reservation = self.get_reservation_by_id(reservation_id)
        reservation.check_in() # Changes the reservation status to ACTIVE and marks the room as occupied. If the reservation is cancelled or completed, it raises an InvalidReservationStatusError to prevent check-in.
        return reservation

    def check_out(self, reservation_id):
        """Check a guest out."""
        reservation = self.get_reservation_by_id(reservation_id)
        reservation.check_out()
        return reservation

    def add_service_to_reservation(self, reservation_id, service_id):
        """Add a service to a reservation."""
        reservation = self.get_reservation_by_id(reservation_id)
        service = self.get_service_by_id(service_id)

        reservation.add_service(service)
        return reservation

    def calculate_accommodation_cost(self, reservation_id):
        """Calculate accommodation cost only."""
        reservation = self.get_reservation_by_id(reservation_id)
        return reservation.calculate_room_charge()

    def calculate_total_charge(self, reservation_id):
        """Calculate total bill including services."""
        reservation = self.get_reservation_by_id(reservation_id)
        return reservation.calculate_total_charge()

    def generate_final_bill(self, reservation_id):
        """Generate final bill summary."""
        reservation = self.get_reservation_by_id(reservation_id)
        return reservation.get_bill_summary()

    def process_payment(self, reservation_id, payment_method, amount=None, **payment_details):
        """
        Process a payment for a reservation.

        Supported payment methods:
        - cash
        - card
        - mobile
        
        **payment_details: Captures method-specific parameters (such as cash_received, card_number, or phone_number) into a dictionary and routes them to the appropriate subclass constructor (CashPayment, CardPayment, or MobilePayment). 
        
        """
        reservation = self.get_reservation_by_id(reservation_id)

        if reservation.status == Reservation.CANCELLED:
            raise InvalidReservationStatusError(
                "Payments cannot be processed for a cancelled reservation."
            )

        if reservation.is_paid:
            raise PaymentError("Reservation is already fully paid.")

        if amount is None:
            amount = reservation.balance_due

        if isinstance(amount, bool) or not isinstance(amount, (int, float)):
            raise PaymentError("Payment amount must be a number.")

        if amount <= 0:
            raise PaymentError("Payment amount must be greater than zero.")

        if amount > reservation.balance_due + 0.01:
            raise PaymentError(
                "Payment amount cannot be greater than the outstanding balance."
            )

        payment_method = str(payment_method).strip().lower()

        if payment_method == "cash":
            payment = CashPayment(
                amount=amount,
                cash_received=payment_details.get("cash_received", amount)
            )

        elif payment_method == "card":
            payment = CardPayment(
                amount=amount,
                card_number=payment_details.get("card_number"),
                expiry_date=payment_details.get("expiry_date"),
                cvv=payment_details.get("cvv")
            )

        elif payment_method == "mobile":
            payment = MobilePayment(
                amount=amount,
                phone_number=payment_details.get("phone_number"),
                provider=payment_details.get("provider")
            )

        else:
            raise PaymentError(
                "Unsupported payment method. Use 'cash', 'card', or 'mobile'."
            )

        payment.process_payment() # Calls the process_payment() method of the specific Payment subclass (CashPayment, CardPayment, or MobilePayment). This method simulates the payment processing and updates the payment status to COMPLETED if successful. If the payment fails (e.g., due to invalid card details or insufficient cash), it raises a PaymentError.
        reservation.record_payment(payment) # Triggers the reservation's record_payment() method to log the completed payment against the reservation. This updates the reservation's payment status and ensures that the payment is accounted for in the overall billing.

        return payment

    def get_payment_history(self, reservation_id):
        """Return all payments made for a reservation."""
        reservation = self.get_reservation_by_id(reservation_id)
        return reservation.payments

    def generate_payment_receipt(self, payment):
        """Generate a receipt for a payment object."""
        if not isinstance(payment, Payment):
            raise TypeError("payment must be an instance of Payment.")

        return payment.get_receipt()

    def get_guest_history(self, guest_id):
        """Return all reservations belonging to a guest."""
        guest = self.get_guest_by_id(guest_id)

        history = [
            reservation for reservation in self._reservations
            if reservation.guest.id_number == guest.id_number
        ]

        return history

    def display_guest_history(self, guest_id):
        """Display all reservations belonging to a guest."""
        history = self.get_guest_history(guest_id)

        if not history:
            return "No reservation history found for this guest."

        return "\n".join(str(reservation) for reservation in history)

    def display_available_rooms(self):
        """Display all available rooms."""
        available_rooms = [
            str(room) for room in self._rooms
            if room.is_available()
        ]

        if not available_rooms:
            return "No available rooms."

        return "\n".join(available_rooms)

    def display_hotel_summary(self):
        """Display hotel summary including revenue information."""
        total_rooms = len(self._rooms)
        available_rooms = len([
            room for room in self._rooms
            if room.is_available()
        ])

        total_guests = len(self._guests)
        total_staff = len(self._staff)
        total_reservations = len(self._reservations)

        active_reservations = len([
            reservation for reservation in self._reservations
            if reservation.status in [Reservation.PENDING, Reservation.ACTIVE]
        ])

        completed_reservations = len([
            reservation for reservation in self._reservations
            if reservation.status == Reservation.COMPLETED
        ])

        cancelled_reservations = len([
            reservation for reservation in self._reservations
            if reservation.status == Reservation.CANCELLED
        ])

        total_revenue = round(sum(
            reservation.amount_paid for reservation in self._reservations
            if reservation.status != Reservation.CANCELLED
        ), 2)

        outstanding_balance = round(sum(
            reservation.balance_due for reservation in self._reservations
            if reservation.status != Reservation.CANCELLED
        ), 2)

        lines = []
        lines.append("=" * 45)
        lines.append(f"HOTEL SUMMARY - {self.name}")
        lines.append("=" * 45)
        lines.append(f"Total Rooms: {total_rooms}")
        lines.append(f"Available Rooms: {available_rooms}")
        lines.append(f"Total Guests: {total_guests}")
        lines.append(f"Total Staff: {total_staff}")
        lines.append(f"Total Reservations: {total_reservations}")
        lines.append(f"Active/Pending Reservations: {active_reservations}")
        lines.append(f"Completed Reservations: {completed_reservations}")
        lines.append(f"Cancelled Reservations: {cancelled_reservations}")
        lines.append(f"Total Revenue: {total_revenue:.2f}")
        lines.append(f"Outstanding Balance: {outstanding_balance:.2f}")
        lines.append("=" * 45)

        return "\n".join(lines)

    def __str__(self):
        return (
            f"Hotel: {self.name} | Rooms: {len(self._rooms)} | "
            f"Guests: {len(self._guests)}"
        )
````

We attempt `make_reservation("CU2026001", "201", -2)`. Expected: `InvalidNightsError`.

````python
try:
    hotel.make_reservation("CU2026001", "201", -2)
except InvalidNightsError as error:
    print("Exception caught:", error)
````

is touched. Validation protects the domain object's invariants.

##### Double booking a reserved room

We attempt `make_reservation("CU2026001", "301", 1)` while room 301 is held by `RES-1002`. Expected: `RoomUnavailableError`.

#### Data Persistence

**Persistence** means converting live objects into a storable form (serialization), writing them to disk, and reconstructing equivalent objects later (deserialization). Without it, every hotel object disappears when the kernel stops.

Strategy used:
- **Reference by ID** for shared entities: a reservation stores `guest_id`,
  `room_number` and `services` IDs, because those objects already exist at the top level of the file. This avoids duplicated, desynchronised copies.
- **Embedding** for owned entities: payments live *inside* their reservation.
- **Polymorphic type tags** (`"class": "Suite"`, `"class": "CardPayment"`)so the correct subclass is rebuilt on load.
- **ISO-format dates** via `.isoformat()` / `datetime.fromisoformat()`.
- **`Class.__new__(Class)`** on load to bypass constructors — re-running
  `Reservation.__init__()` would mint a new ID and re-reserve an already-held room.
- **Counter resynchronisation** so newly created objects never reuse loaded IDs.
- **Card safety:** only `card_last4` and expiry are written; full PAN and CVV never are.

The system can now:

- Save hotel data to a JSON file
- Load hotel data from a JSON file
- Preserve guests, staff, rooms, services, reservations, and payments
- Maintain the hotel state after saving and loading

````python
# Persistence Imports and Data Folder

import json
import os

DATA_DIRECTORY = "data"
DATA_FILE = os.path.join(DATA_DIRECTORY, "hotel_data.json")

os.makedirs(DATA_DIRECTORY, exist_ok=True)
````

#####  Object-to-Dictionary Converters

````python
def room_to_dict(room):
    """Convert a Room object into a dictionary."""
    room_class = room.__class__.__name__

    data = {
        "class": room_class,
        "room_number": room.room_number,
        "base_price": room.base_price,
        "status": room.status
    }

    if isinstance(room, StandardRoom):
        data["has_tv"] = room.has_tv

    elif isinstance(room, DeluxeRoom):
        data["includes_breakfast"] = room.includes_breakfast

    elif isinstance(room, Suite):
        data["lounge_access"] = room.lounge_access

    return data


def service_to_dict(service):
    """Convert a HotelService object into a dictionary."""
    return {
        "service_id": service.service_id,
        "name": service.name,
        "price": service.price
    }


def payment_to_dict(payment):
    """Convert a Payment object into a dictionary."""
    data = {
        "class": payment.__class__.__name__,
        "payment_id": payment.payment_id,
        "amount": payment.amount,
        "payment_date": payment.payment_date.isoformat(),
        "status": payment.status
    }

    if isinstance(payment, CashPayment):
        data["cash_received"] = payment.cash_received

    elif isinstance(payment, CardPayment):
        data["card_last4"] = payment.masked_card_number[-4:]
        data["expiry_date"] = payment._expiry_date

    elif isinstance(payment, MobilePayment):
        data["phone_number"] = payment.phone_number
        data["provider"] = payment.provider

    return data


def reservation_to_dict(reservation):
    """Convert a Reservation object into a dictionary."""
    return {
        "reservation_id": reservation.reservation_id,
        "guest_id": reservation.guest.id_number,
        "room_number": reservation.room.room_number,
        "nights": reservation.nights,
        "check_in_date": reservation.check_in_date.isoformat(),
        "status": reservation.status,
        "payment_status": reservation.payment_status,
        "created_on": reservation.created_on.isoformat(),
        "services": [
            service.service_id for service in reservation.services
        ],
        "payments": [
            payment_to_dict(payment) for payment in reservation.payments
        ]
    }


def guest_to_dict(guest):
    """Convert a Guest object into a dictionary."""
    return {
        "name": guest.name,
        "id_number": guest.id_number,
        "phone": guest.phone,
        "guest_type": guest.guest_type,
        "registered_on": guest.registered_on.isoformat(),
        "loyalty_points": guest.loyalty_points
    }


def staff_to_dict(staff):
    """Convert a Staff object into a dictionary."""
    return {
        "name": staff.name,
        "id_number": staff.id_number,
        "role": staff.role
    }


def hotel_to_dict(hotel):
    """Convert the whole Hotel object into a dictionary."""
    return {
        "hotel": {
            "name": hotel.name
        },
        "guests": [
            guest_to_dict(guest) for guest in hotel.guests
        ],
        "staff": [
            staff_to_dict(staff) for staff in hotel.staff
        ],
        "rooms": [
            room_to_dict(room) for room in hotel.rooms
        ],
        "services": [
            service_to_dict(service) for service in hotel.services
        ],
        "reservations": [
            reservation_to_dict(reservation)
            for reservation in hotel.reservations
        ]
    }
````

````python
def save_hotel(hotel, filepath=DATA_FILE):
    """
    Save the hotel system to a JSON file atomically.
    Writing to a temporary file and then os.replace() guarantees the data file is
    never left half-written if the process crashes mid-save.
    """
    directory = os.path.dirname(filepath)
    if directory:
        os.makedirs(directory, exist_ok=True)

    hotel_data = hotel_to_dict(hotel)
    hotel_data["version"] = 1

    temp_path = filepath + ".tmp"
    with open(temp_path, "w") as file:
        json.dump(hotel_data, file, indent=4)
    os.replace(temp_path, filepath)

    return filepath
````

##### Dictionary-to-Object Converters

````python
def payment_from_dict(payment_data):
    """Recreate a Payment object from a dictionary."""
    payment_class = payment_data.get("class")

    if payment_class == "CashPayment":
        payment = CashPayment.__new__(CashPayment)

        payment._payment_id = payment_data["payment_id"]
        payment._amount = float(payment_data["amount"])
        payment._payment_date = datetime.fromisoformat(payment_data["payment_date"])
        payment._status = payment_data["status"]
        payment._cash_received = float(payment_data.get("cash_received", payment_data["amount"]))

        return payment

    elif payment_class == "CardPayment":
        payment = CardPayment.__new__(CardPayment)

        payment._payment_id = payment_data["payment_id"]
        payment._amount = float(payment_data["amount"])
        payment._payment_date = datetime.fromisoformat(payment_data["payment_date"])
        payment._status = payment_data["status"]

        card_last4 = payment_data.get("card_last4", "0000")
        payment._card_number = ("0000000000000000" + card_last4)[-16:]
        payment._expiry_date = payment_data.get("expiry_date", "12/99")
        payment._cvv = "000"

        return payment

    elif payment_class == "MobilePayment":
        payment = MobilePayment.__new__(MobilePayment)

        payment._payment_id = payment_data["payment_id"]
        payment._amount = float(payment_data["amount"])
        payment._payment_date = datetime.fromisoformat(payment_data["payment_date"])
        payment._status = payment_data["status"]
        payment._phone_number = payment_data.get("phone_number", "")
        payment._provider = payment_data.get("provider", "")

        return payment

    else:
        raise ValueError(f"Unknown payment class: {payment_class}")


def reservation_from_dict(hotel, reservation_data, guest_map, room_map, service_map):
    """Recreate a Reservation object from a dictionary."""
    guest = guest_map[reservation_data["guest_id"]]
    room = room_map[reservation_data["room_number"]]

    reservation = Reservation.__new__(Reservation)

    reservation._reservation_id = reservation_data["reservation_id"]
    reservation._guest = guest
    reservation._room = room
    reservation._nights = reservation_data["nights"]
    reservation._check_in_date = datetime.fromisoformat(
        reservation_data["check_in_date"]
    ).date()
    reservation._status = reservation_data["status"]
    reservation._payment_status = reservation_data["payment_status"]
    reservation._created_on = datetime.fromisoformat(reservation_data["created_on"])

    reservation._services = []

    for service_id in reservation_data.get("services", []):
        if service_id in service_map:
            reservation._services.append(service_map[service_id])

    reservation._payments = []

    for payment_data in reservation_data.get("payments", []):
        payment = payment_from_dict(payment_data)
        reservation._payments.append(payment)

    return reservation
````

##### Load Hotel Function

````python
def update_next_id(object_ids, prefix, target_class, attribute_name):
    """
    Resynchronise a class ID counter after loading.

    Monotonic: the counter is only ever raised, never lowered. Lowering it to the
    loaded file's maximum could re-issue IDs that were already created earlier in
    this session (e.g. by the test hotels).
    """
    maximum_value = 0
    for object_id in object_ids:
        try:
            value = int(str(object_id).split(prefix)[1])
        except (ValueError, IndexError):
            continue
        if value > maximum_value:
            maximum_value = value

    if maximum_value > getattr(target_class, attribute_name):
        setattr(target_class, attribute_name, maximum_value)


def load_hotel(filepath=DATA_FILE):
    """Load the hotel system from a JSON file, restoring objects and ID counters."""
    if not os.path.exists(filepath):
        raise FileNotFoundError(f"Data file not found: {filepath}")

    with open(filepath, "r") as file:
        data = json.load(file)

    hotel = Hotel(data["hotel"]["name"])
    guest_map, room_map, service_map = {}, {}, {}

    for guest_data in data.get("guests", []):
        guest = Guest(
            name=guest_data["name"],
            id_number=guest_data["id_number"],
            phone=guest_data["phone"],
            guest_type=guest_data.get("guest_type", "Regular"),
        )
        guest.registered_on = datetime.fromisoformat(guest_data["registered_on"]).date()
        guest._loyalty_points = guest_data.get("loyalty_points", 0)
        hotel._guests.append(guest)
        guest_map[guest.id_number] = guest

    Guest.guest_count = len(hotel._guests)   # set once, after the loop

    for staff_data in data.get("staff", []):
        hotel._staff.append(Staff(
            name=staff_data["name"],
            id_number=staff_data["id_number"],
            role=staff_data["role"],
        ))

    for room_data in data.get("rooms", []):
        room_class = room_data.get("class")
        if room_class == "StandardRoom":
            room = StandardRoom(room_data["room_number"], room_data["base_price"],
                                has_tv=room_data.get("has_tv", True))
        elif room_class == "DeluxeRoom":
            room = DeluxeRoom(room_data["room_number"], room_data["base_price"],
                              includes_breakfast=room_data.get("includes_breakfast", True))
        elif room_class == "Suite":
            room = Suite(room_data["room_number"], room_data["base_price"],
                         lounge_access=room_data.get("lounge_access", True))
        else:
            raise ValueError(f"Unknown room class: {room_class}")
        room._status = room_data.get("status", Room.AVAILABLE)
        hotel._rooms.append(room)
        room_map[room.room_number] = room

    for service_data in data.get("services", []):
        service = HotelService(service_data["name"], service_data["price"])
        service.service_id = service_data["service_id"]
        hotel._services.append(service)
        service_map[service.service_id] = service

    for reservation_data in data.get("reservations", []):
        hotel._reservations.append(reservation_from_dict(
            hotel=hotel,
            reservation_data=reservation_data,
            guest_map=guest_map,
            room_map=room_map,
            service_map=service_map,
        ))

    update_next_id([s.service_id for s in hotel.services], "SRV-",
                   HotelService, "_next_service_id")
    update_next_id([p.payment_id for r in hotel.reservations for p in r.payments],
                   "PAY-", Payment, "_next_payment_id")
    update_next_id([r.reservation_id for r in hotel.reservations], "RES-",
                   Reservation, "_next_reservation_id")

    return hotel
````

```text
This section provides a desktop Graphical User Interface for the hotel system using Python's built-in `tkinter` library.

````python
import os
import tkinter as tk
from tkinter import ttk, messagebox
from tkinter import font as tkfont


class HotelGUI:
    """
    Desktop GUI for the Hotel Reservation and Management System.
    """

    # ------------------------------------------------------------------ #
    # Setup
    # ------------------------------------------------------------------ #
    def __init__(self, hotel):
        self.hotel = hotel

        self.root = tk.Tk()
        self.root.title(f"{self.hotel.name} - Hotel Management System")
        self.root.geometry("1220x760")
        self.root.minsize(1060, 640)

        self.setup_styles()
        self.build_header()
        self.build_toolbar()
        self.build_status_bar()
        self.create_layout()
        self.refresh_all()

        # Offer to restore persisted state on startup
        if os.path.exists(DATA_FILE):
            if messagebox.askyesno(
                "Load Saved Data",
                f"Saved data found:\n{DATA_FILE}\n\nLoad it now?",
                parent=self.root,
            ):
                self.load_data(confirm=False)

    def run(self):
        """Start the GUI main loop (blocks until the window is closed)."""
        self.root.mainloop()

    def setup_styles(self):
        """Apply one consistent, larger-font theme to every widget."""
        available = set(tkfont.families(self.root))
        ui = next((f for f in ("Segoe UI", "Helvetica Neue", "DejaVu Sans", "Arial")
                   if f in available), "TkDefaultFont")
        mono = next((f for f in ("Consolas", "Menlo", "DejaVu Sans Mono", "Courier New")
                     if f in available), "TkFixedFont")
        self.mono_font = (mono, 11)

        style = ttk.Style(self.root)
        style.theme_use("clam")
        style.configure(".", font=(ui, 11))
        style.configure("TButton", font=(ui, 10), padding=(10, 6))
        style.configure("TNotebook.Tab", font=(ui, 11, "bold"), padding=(16, 8))
        style.configure("Treeview", font=(ui, 11), rowheight=30, fieldbackground="white")
        style.configure("Treeview.Heading", font=(ui, 11, "bold"), padding=(6, 8))
        style.map("Treeview",
                  background=[("selected", "#2563eb")],
                  foreground=[("selected", "white")])
        style.configure("Header.TLabel", font=(ui, 24, "bold"), foreground="#111827")
        style.configure("SubHeader.TLabel", font=(ui, 10), foreground="#6b7280")
        style.configure("Card.TFrame", background="white", relief="solid", borderwidth=1)
        style.configure("CardTitle.TLabel", font=(ui, 9, "bold"),
                        background="white", foreground="#6b7280")
        style.configure("CardValue.TLabel", font=(ui, 20, "bold"),
                        background="white", foreground="#111827")

    def build_header(self):
        header = ttk.Frame(self.root, padding=(18, 14, 18, 8))
        header.pack(fill="x")
        self.title_label = ttk.Label(header, text=self.hotel.name, style="Header.TLabel")
        self.title_label.pack(anchor="w")
        ttk.Label(header, text="Hotel Reservation & Management System",
                  style="SubHeader.TLabel").pack(anchor="w")
        ttk.Separator(self.root).pack(fill="x")

    def build_toolbar(self):
        bar = ttk.Frame(self.root, padding=(18, 8))
        bar.pack(fill="x")
        ttk.Button(bar, text="Save Data", command=self.save_data).pack(side="left")
        ttk.Button(bar, text="Load Data", command=self.load_data).pack(side="left", padx=8)

    def build_status_bar(self):
        self.status_var = tk.StringVar(value="Ready")
        ttk.Label(self.root, textvariable=self.status_var,
                  style="SubHeader.TLabel", padding=(18, 4)).pack(side="bottom", anchor="w")

    def create_layout(self):
        self.notebook = ttk.Notebook(self.root)
        self.notebook.pack(fill="both", expand=True, padx=10, pady=(0, 10))

        self.dashboard_tab = ttk.Frame(self.notebook)
        self.rooms_tab = ttk.Frame(self.notebook)
        self.people_tab = ttk.Frame(self.notebook)
        self.services_tab = ttk.Frame(self.notebook)
        self.reservations_tab = ttk.Frame(self.notebook)

        self.notebook.add(self.dashboard_tab, text="Dashboard")
        self.notebook.add(self.rooms_tab, text="Rooms")
        self.notebook.add(self.people_tab, text="Guests & Staff")
        self.notebook.add(self.services_tab, text="Services")
        self.notebook.add(self.reservations_tab, text="Reservations & Payments")

        self.build_dashboard()
        self.build_rooms()
        self.build_people()
        self.build_services()
        self.build_reservations()

    # ------------------------------------------------------------------ #
    # Reusable helpers
    # ------------------------------------------------------------------ #
    def make_tree(self, parent, columns, headings, widths):
        """Create a styled, scrollable table with zebra-stripe tags."""
        frame = ttk.Frame(parent)
        frame.pack(fill="both", expand=True, padx=4, pady=4)
        tree = ttk.Treeview(frame, columns=columns, show="headings")
        for col, text, width in zip(columns, headings, widths):
            tree.heading(col, text=text)
            tree.column(col, width=width, anchor="w")
        bar = ttk.Scrollbar(frame, orient="vertical", command=tree.yview)
        tree.configure(yscrollcommand=bar.set)
        bar.pack(side="right", fill="y")
        tree.pack(side="left", fill="both", expand=True)
        tree.tag_configure("odd", background="#f3f4f6")
        tree.tag_configure("even", background="white")
        return tree

    def ask_form(self, title, fields):
        """
        Modal form dialog. Replaces all chained simpledialog popups.

        fields: list of dicts with keys:
          key, label, kind ('text'|'int'|'float'|'choice'),
          default, options (for choice), optional (bool), show (e.g. '*')
        Returns a dict of values, or None if cancelled.
        """
        dialog = tk.Toplevel(self.root)
        dialog.title(title)
        dialog.transient(self.root)
        dialog.grab_set()
        dialog.resizable(False, False)

        widgets = []
        for row, field in enumerate(fields):
            ttk.Label(dialog, text=field["label"] + ":").grid(
                row=row, column=0, sticky="w", padx=(16, 8), pady=6)
            kind = field.get("kind", "text")
            default = field.get("default", "")
            if kind == "choice":
                var = tk.StringVar(value=str(default) if default else field["options"][0])
                widget = ttk.Combobox(dialog, textvariable=var,
                                      values=field["options"], state="readonly", width=30)
            else:
                var = tk.StringVar(value=str(default))
                widget = ttk.Entry(dialog, textvariable=var, width=32,
                                   show=field.get("show", ""))
            widget.grid(row=row, column=1, padx=(0, 16), pady=6)
            widgets.append((field, var, widget))

        result = {"data": None}

        def on_ok(event=None):
            data = {}
            for field, var, _ in widgets:
                raw = var.get().strip()
                key, kind = field["key"], field.get("kind", "text")
                if raw == "":
                    if field.get("optional"):
                        data[key] = None
                        continue
                    messagebox.showerror("Missing Input",
                                         f"{field['label']} is required.", parent=dialog)
                    return
                if kind == "int":
                    try:
                        data[key] = int(raw)
                    except ValueError:
                        messagebox.showerror("Invalid Input",
                                             f"{field['label']} must be a whole number.",
                                             parent=dialog)
                        return
                elif kind == "float":
                    try:
                        data[key] = float(raw)
                    except ValueError:
                        messagebox.showerror("Invalid Input",
                                             f"{field['label']} must be a number.",
                                             parent=dialog)
                        return
                else:
                    data[key] = raw
            result["data"] = data
            dialog.destroy()

        def on_cancel(event=None):
            dialog.destroy()

        buttons = ttk.Frame(dialog)
        buttons.grid(row=len(fields), column=0, columnspan=2, pady=(8, 14))
        ttk.Button(buttons, text="OK", command=on_ok).pack(side="left", padx=6)
        ttk.Button(buttons, text="Cancel", command=on_cancel).pack(side="left")

        dialog.bind("<Return>", on_ok)
        dialog.bind("<Escape>", on_cancel)
        dialog.protocol("WM_DELETE_WINDOW", on_cancel)
        if widgets:
            widgets[0][2].focus_set()

        dialog.update_idletasks()
        x = self.root.winfo_rootx() + (self.root.winfo_width() - dialog.winfo_width()) // 2
        y = self.root.winfo_rooty() + 90
        dialog.geometry(f"+{max(x, 0)}+{max(y, 0)}")
        self.root.wait_window(dialog)
        return result["data"]

    def show_document(self, title, content):
        """Bills / receipts / history in a monospaced, scrollable, read-only window."""
        win = tk.Toplevel(self.root)
        win.title(title)
        win.geometry("620x660")
        win.transient(self.root)
        text = tk.Text(win, font=self.mono_font, wrap="none",
                       padx=14, pady=14, background="#fafafa")
        bar = ttk.Scrollbar(win, orient="vertical", command=text.yview)
        text.configure(yscrollcommand=bar.set)
        bar.pack(side="right", fill="y")
        text.pack(fill="both", expand=True)
        text.insert("1.0", content)
        text.configure(state="disabled")

    def get_selected_reservation_id(self):
        selected = self.reservations_tree.selection()
        if not selected:
            messagebox.showwarning("No Selection", "Please select a reservation first.",
                                   parent=self.root)
            return None
        return str(self.reservations_tree.item(selected[0])["values"][0])

    def get_selected_guest_id(self):
        selected = self.guests_tree.selection()
        if not selected:
            messagebox.showwarning("No Selection", "Please select a guest first.",
                                   parent=self.root)
            return None
        return str(self.guests_tree.item(selected[0])["values"][0])

    def run_reservation_action(self, operation, success_message):
        """Shared wrapper for check-in / check-out / cancel."""
        reservation_id = self.get_selected_reservation_id()
        if not reservation_id:
            return
        try:
            operation(reservation_id)
            self.refresh_all()
            self.status_var.set(f"{reservation_id}: {success_message}")
            messagebox.showinfo("Success", f"{reservation_id}: {success_message}",
                                parent=self.root)
        except Exception as error:
            messagebox.showerror("Error", str(error), parent=self.root)

    # ------------------------------------------------------------------ #
    # Dashboard tab
    # ------------------------------------------------------------------ #
    def build_dashboard(self):
        cards = ttk.Frame(self.dashboard_tab)
        cards.pack(fill="x", padx=8, pady=(10, 4))
        self.stat_labels = {}
        definitions = [("TOTAL ROOMS", "rooms"), ("AVAILABLE", "available"),
                       ("GUESTS", "guests"), ("REVENUE", "revenue"),
                       ("OUTSTANDING", "outstanding")]
        for col, (title, key) in enumerate(definitions):
            cards.columnconfigure(col, weight=1)
            card = ttk.Frame(cards, style="Card.TFrame", padding=14)
            card.grid(row=0, column=col, padx=6, sticky="nsew")
            ttk.Label(card, text=title, style="CardTitle.TLabel").pack(anchor="w")
            label = ttk.Label(card, text="—", style="CardValue.TLabel")
            label.pack(anchor="w")
            self.stat_labels[key] = label

        bar = ttk.Frame(self.dashboard_tab)
        bar.pack(fill="x", padx=8, pady=4)
        ttk.Button(bar, text="Refresh Dashboard",
                   command=self.refresh_dashboard).pack(side="left")

        self.dashboard_text = tk.Text(self.dashboard_tab, font=self.mono_font,
                                      padx=14, pady=12, relief="flat", background="#fafafa")
        self.dashboard_text.pack(fill="both", expand=True, padx=8, pady=(0, 8))
        self.dashboard_text.tag_configure(
            "heading", font=(self.mono_font[0], 14, "bold"),
            foreground="#111827", spacing1=6, spacing3=6)
        self.dashboard_text.tag_configure(
            "section", font=(self.mono_font[0], 12, "bold"),
            foreground="#2563eb", spacing1=10)

    def refresh_dashboard(self):
        rooms = self.hotel.rooms
        live = [r for r in self.hotel.reservations if r.status != "CANCELLED"]
        stats = {
            "rooms": str(len(rooms)),
            "available": str(sum(1 for r in rooms if r.is_available())),
            "guests": str(len(self.hotel.guests)),
            "revenue": f"{sum(r.amount_paid for r in live):,.2f}",
            "outstanding": f"{sum(r.balance_due for r in live):,.2f}",
        }
        for key, value in stats.items():
            self.stat_labels[key].configure(text=value)

        self.dashboard_text.configure(state="normal")
        self.dashboard_text.delete("1.0", tk.END)
        self.dashboard_text.insert(tk.END, "Hotel Summary\n", "heading")
        self.dashboard_text.insert(tk.END, self.hotel.display_hotel_summary() + "\n")
        # self.dashboard_text.insert(tk.END, "Available Rooms\n", "section")
        # self.dashboard_text.insert(tk.END, self.hotel.display_available_rooms())
        self.dashboard_text.configure(state="disabled")
    
  
    # ------------------------------------------------------------------ #
    # Rooms tab
    # ------------------------------------------------------------------ #
    def build_rooms(self):
        bar = ttk.Frame(self.rooms_tab)
        bar.pack(fill="x", padx=8, pady=8)
        ttk.Button(bar, text="Add Room", command=self.add_room_dialog).pack(side="left")
        ttk.Button(bar, text="Search Rooms",
                   command=self.search_rooms_dialog).pack(side="left", padx=6)
        ttk.Button(bar, text="Show All / Refresh",
                   command=self.refresh_rooms).pack(side="left")

        self.rooms_tree = self.make_tree(
            self.rooms_tab,
            ("number", "type", "status", "price"),
            ("Room Number", "Room Type", "Status", "Base Price"),
            (140, 240, 160, 160),
        )
        self.rooms_tree.column("price", anchor="e")
        self.rooms_tree.tag_configure("AVAILABLE", foreground="#047857")
        self.rooms_tree.tag_configure("RESERVED", foreground="#b45309")
        self.rooms_tree.tag_configure("OCCUPIED", foreground="#b91c1c")
        self.rooms_tree.tag_configure("CLEANING", foreground="#6b7280")
        self.rooms_tree.tag_configure("MAINTENANCE", foreground="#6b7280")

    def refresh_rooms(self, rooms=None):
        rooms = self.hotel.rooms if rooms is None else rooms
        for item in self.rooms_tree.get_children():
            self.rooms_tree.delete(item)
        for i, room in enumerate(rooms):
            self.rooms_tree.insert(
                "", tk.END,
                values=(room.room_number, room.__class__.__name__,
                        room.status, f"{room.base_price:,.2f}"),
                tags=("odd" if i % 2 else "even", room.status),
            )

    def add_room_dialog(self):
        data = self.ask_form("Add Room", [
            {"key": "room_type", "label": "Room type", "kind": "choice",
             "options": ["standard", "deluxe", "suite"]},
            {"key": "room_number", "label": "Room number"},
            {"key": "base_price", "label": "Base price per night", "kind": "float"},
        ])
        if data is None:
            return
        try:
            room = self.hotel.add_room(**data)
            self.refresh_all()
            self.status_var.set(f"Room {room.room_number} added.")
        except Exception as error:
            messagebox.showerror("Error", str(error), parent=self.root)

    def search_rooms_dialog(self):
        data = self.ask_form("Search Rooms", [
            {"key": "room_number", "label": "Room number (blank = any)", "optional": True},
            {"key": "room_type", "label": "Room type", "kind": "choice",
             "options": ["any", "standard", "deluxe", "suite"]},
            {"key": "available_only", "label": "Available only?", "kind": "choice",
             "options": ["yes", "no"]},
        ])
        if data is None:
            return
        try:
            rooms = self.hotel.search_rooms(
                room_number=data["room_number"],
                room_type=None if data["room_type"] == "any" else data["room_type"],
                available_only=data["available_only"] == "yes",
            )
            self.refresh_rooms(rooms)
            self.status_var.set(f"Search complete: {len(rooms)} room(s) found.")
        except Exception as error:
            messagebox.showerror("Error", str(error), parent=self.root)

    # ------------------------------------------------------------------ #
    # Guests & Staff tab
    # ------------------------------------------------------------------ #
    def build_people(self):
        bar = ttk.Frame(self.people_tab)
        bar.pack(fill="x", padx=8, pady=8)
        ttk.Button(bar, text="Register Guest",
                   command=self.register_guest_dialog).pack(side="left")
        ttk.Button(bar, text="Register Staff",
                   command=self.register_staff_dialog).pack(side="left", padx=6)
        ttk.Button(bar, text="Guest History",
                   command=self.guest_history_action).pack(side="left")
        ttk.Button(bar, text="Refresh",
                   command=self.refresh_people).pack(side="left", padx=6)

        container = ttk.Frame(self.people_tab)
        container.pack(fill="both", expand=True, padx=8, pady=(0, 8))
        guest_frame = ttk.LabelFrame(container, text="Guests", padding=4)
        guest_frame.pack(side="left", fill="both", expand=True)
        staff_frame = ttk.LabelFrame(container, text="Staff", padding=4)
        staff_frame.pack(side="left", fill="both", expand=True, padx=(8, 0))

        self.guests_tree = self.make_tree(
            guest_frame,
            ("id", "name", "phone", "type"),
            ("Guest ID", "Name", "Phone", "Type"),
            (130, 170, 150, 90),
        )
        self.staff_tree = self.make_tree(
            staff_frame,
            ("id", "name", "role"),
            ("Staff ID", "Name", "Role"),
            (130, 180, 140),
        )

    def refresh_people(self):
        for item in self.guests_tree.get_children():
            self.guests_tree.delete(item)
        for i, guest in enumerate(self.hotel.guests):
            self.guests_tree.insert(
                "", tk.END,
                values=(guest.id_number, guest.name, guest.phone, guest.guest_type),
                tags=("odd" if i % 2 else "even",),
            )
        for item in self.staff_tree.get_children():
            self.staff_tree.delete(item)
        for i, staff in enumerate(self.hotel.staff):
            self.staff_tree.insert(
                "", tk.END,
                values=(staff.id_number, staff.name, staff.role),
                tags=("odd" if i % 2 else "even",),
            )

    def register_guest_dialog(self):
        data = self.ask_form("Register Guest", [
            {"key": "name", "label": "Guest name"},
            {"key": "id_number", "label": "Guest ID"},
            {"key": "phone", "label": "Phone number"},
            {"key": "guest_type", "label": "Guest type", "kind": "choice",
             "options": ["Regular", "VIP"]},
        ])
        if data is None:
            return
        try:
            guest = self.hotel.register_guest(**data)
            self.refresh_all()
            self.status_var.set(f"Guest {guest.name} registered.")
        except Exception as error:
            messagebox.showerror("Error", str(error), parent=self.root)

    def register_staff_dialog(self):
        data = self.ask_form("Register Staff", [
            {"key": "name", "label": "Staff name"},
            {"key": "id_number", "label": "Staff ID"},
            {"key": "role", "label": "Role", "default": "Receptionist"},
        ])
        if data is None:
            return
        try:
            staff = self.hotel.register_staff(**data)
            self.refresh_all()
            self.status_var.set(f"Staff {staff.name} registered.")
        except Exception as error:
            messagebox.showerror("Error", str(error), parent=self.root)

    def guest_history_action(self):
        guest_id = self.get_selected_guest_id()
        if not guest_id:
            return
        try:
            content = self.hotel.display_guest_history(guest_id)
            self.show_document(f"Guest History - {guest_id}", content)
        except Exception as error:
            messagebox.showerror("Error", str(error), parent=self.root)

    # ------------------------------------------------------------------ #
    # Services tab
    # ------------------------------------------------------------------ #
    def build_services(self):
        bar = ttk.Frame(self.services_tab)
        bar.pack(fill="x", padx=8, pady=8)
        ttk.Button(bar, text="Add Service",
                   command=self.add_service_dialog).pack(side="left")
        ttk.Button(bar, text="Refresh",
                   command=self.refresh_services).pack(side="left", padx=6)

        self.services_tree = self.make_tree(
            self.services_tab,
            ("id", "name", "price"),
            ("Service ID", "Name", "Price"),
            (160, 320, 160),
        )
        self.services_tree.column("price", anchor="e")

    def refresh_services(self):
        for item in self.services_tree.get_children():
            self.services_tree.delete(item)
        for i, service in enumerate(self.hotel.services):
            self.services_tree.insert(
                "", tk.END,
                values=(service.service_id, service.name, f"{service.price:,.2f}"),
                tags=("odd" if i % 2 else "even",),
            )

    def add_service_dialog(self):
        data = self.ask_form("Add Service", [
            {"key": "name", "label": "Service name"},
            {"key": "price", "label": "Price", "kind": "float"},
        ])
        if data is None:
            return
        try:
            service = self.hotel.add_service(**data)
            self.refresh_all()
            self.status_var.set(f"Service {service.service_id} added.")
        except Exception as error:
            messagebox.showerror("Error", str(error), parent=self.root)

    # ------------------------------------------------------------------ #
    # Reservations & Payments tab
    # ------------------------------------------------------------------ #
    def build_reservations(self):
        row1 = ttk.Frame(self.reservations_tab)
        row1.pack(fill="x", padx=8, pady=(8, 2))
        ttk.Button(row1, text="Make Reservation",
                   command=self.make_reservation_dialog).pack(side="left")
        ttk.Button(row1, text="Add Service",
                   command=self.add_service_action).pack(side="left", padx=6)
        ttk.Button(row1, text="Check In", command=self.check_in_action).pack(side="left")
        ttk.Button(row1, text="Check Out", command=self.check_out_action).pack(side="left", padx=6)
        ttk.Button(row1, text="Cancel", command=self.cancel_action).pack(side="left")

        row2 = ttk.Frame(self.reservations_tab)
        row2.pack(fill="x", padx=8, pady=(2, 8))
        ttk.Button(row2, text="Final Bill", command=self.show_final_bill).pack(side="left")
        ttk.Button(row2, text="Process Payment",
                   command=self.process_payment_dialog).pack(side="left", padx=6)
        ttk.Button(row2, text="Payment History",
                   command=self.payment_history_action).pack(side="left")
        ttk.Button(row2, text="Refresh",
                   command=self.refresh_reservations).pack(side="left", padx=6)

        self.reservations_tree = self.make_tree(
            self.reservations_tab,
            ("id", "guest", "room", "nights", "status", "payment", "total"),
            ("Reservation ID", "Guest", "Room", "Nights", "Status", "Payment", "Total"),
            (130, 200, 80, 70, 120, 150, 130),
        )
        self.reservations_tree.column("total", anchor="e")
        self.reservations_tree.tag_configure("PENDING", foreground="#b45309")
        self.reservations_tree.tag_configure("ACTIVE", foreground="#1d4ed8")
        self.reservations_tree.tag_configure("COMPLETED", foreground="#047857")
        self.reservations_tree.tag_configure("CANCELLED", foreground="#b91c1c")

    def refresh_reservations(self):
        for item in self.reservations_tree.get_children():
            self.reservations_tree.delete(item)
        for i, reservation in enumerate(self.hotel.reservations):
            self.reservations_tree.insert(
                "", tk.END,
                values=(reservation.reservation_id, reservation.guest.name,
                        reservation.room.room_number, reservation.nights,
                        reservation.status, reservation.payment_status,
                        f"{reservation.calculate_total_charge():,.2f}"),
                tags=("odd" if i % 2 else "even", reservation.status),
            )

    def make_reservation_dialog(self):
        if not self.hotel.guests:
            messagebox.showwarning("No Guests", "Register a guest first.", parent=self.root)
            return
        available = [r.room_number for r in self.hotel.rooms if r.is_available()]
        if not available:
            messagebox.showwarning("No Rooms", "No available rooms to book.", parent=self.root)
            return
        data = self.ask_form("Make Reservation", [
            {"key": "guest_id", "label": "Guest ID", "kind": "choice",
             "options": [g.id_number for g in self.hotel.guests]},
            {"key": "room_number", "label": "Available room", "kind": "choice",
             "options": available},
            {"key": "nights", "label": "Number of nights", "kind": "int", "default": 1},
        ])
        if data is None:
            return
        try:
            reservation = self.hotel.make_reservation(**data)
            self.refresh_all()
            self.status_var.set(f"{reservation.reservation_id} created.")
            messagebox.showinfo("Success",
                                f"Reservation {reservation.reservation_id} created.",
                                parent=self.root)
        except Exception as error:
            messagebox.showerror("Error", str(error), parent=self.root)

    def add_service_action(self):
        reservation_id = self.get_selected_reservation_id()
        if not reservation_id:
            return
        if not self.hotel.services:
            messagebox.showwarning("No Services", "Add a hotel service first.", parent=self.root)
            return
        data = self.ask_form("Add Service to Reservation", [
            {"key": "service", "label": "Service", "kind": "choice",
             "options": [f"{s.service_id}  ({s.name} - {s.price:,.2f})"
                         for s in self.hotel.services]},
        ])
        if data is None:
            return
        service_id = data["service"].split()[0]
        try:
            self.hotel.add_service_to_reservation(reservation_id, service_id)
            self.refresh_all()
            self.status_var.set(f"{service_id} added to {reservation_id}.")
        except Exception as error:
            messagebox.showerror("Error", str(error), parent=self.root)

    def check_in_action(self):
        self.run_reservation_action(self.hotel.check_in, "check-in successful.")

    def check_out_action(self):
        self.run_reservation_action(self.hotel.check_out, "check-out successful.")

    def cancel_action(self):
        reservation_id = self.get_selected_reservation_id()
        if not reservation_id:
            return
        if not messagebox.askyesno("Confirm Cancel",
                                   f"Cancel reservation {reservation_id}?",
                                   parent=self.root):
            return
        self.run_reservation_action(self.hotel.cancel_reservation, "reservation cancelled.")

    def show_final_bill(self):
        reservation_id = self.get_selected_reservation_id()
        if not reservation_id:
            return
        try:
            bill = self.hotel.generate_final_bill(reservation_id)
            self.show_document(f"Final Bill - {reservation_id}", bill)
        except Exception as error:
            messagebox.showerror("Error", str(error), parent=self.root)

    def payment_history_action(self):
        reservation_id = self.get_selected_reservation_id()
        if not reservation_id:
            return
        try:
            payments = self.hotel.get_payment_history(reservation_id)
            if not payments:
                messagebox.showinfo("Payment History",
                                    "No payments recorded for this reservation.",
                                    parent=self.root)
                return
            content = f"PAYMENT HISTORY - {reservation_id}\n\n"
            content += "\n".join(str(payment) for payment in payments)
            self.show_document(f"Payment History - {reservation_id}", content)
        except Exception as error:
            messagebox.showerror("Error", str(error), parent=self.root)

    def process_payment_dialog(self):
        reservation_id = self.get_selected_reservation_id()
        if not reservation_id:
            return
        try:
            reservation = self.hotel.get_reservation_by_id(reservation_id)
            base = self.ask_form("Process Payment", [
                {"key": "method", "label": "Payment method", "kind": "choice",
                 "options": ["cash", "card", "mobile"]},
                {"key": "amount",
                 "label": f"Amount (blank = full balance {reservation.balance_due:,.2f})",
                 "kind": "float", "optional": True},
            ])
            if base is None:
                return

            method = base["method"]
            if method == "cash":
                extra = self.ask_form("Cash Details", [
                    {"key": "cash_received",
                     "label": "Cash received (blank = exact amount)",
                     "kind": "float", "optional": True},
                ])
            elif method == "card":
                extra = self.ask_form("Card Details", [
                    {"key": "card_number", "label": "Card number"},
                    {"key": "expiry_date", "label": "Expiry date (MM/YY)"},
                    {"key": "cvv", "label": "CVV", "show": "*"},
                ])
            else:
                extra = self.ask_form("Mobile Money Details", [
                    {"key": "phone_number", "label": "Phone number"},
                    {"key": "provider", "label": "Provider", "default": "M-Pesa"},
                ])
            if extra is None:
                return

            details = {k: v for k, v in extra.items() if v is not None}
            payment = self.hotel.process_payment(
                reservation_id, method, amount=base["amount"], **details)
            self.refresh_all()
            self.status_var.set(f"{payment.payment_id} completed.")
            self.show_document(f"Receipt - {payment.payment_id}", payment.get_receipt())
        except Exception as error:
            messagebox.showerror("Payment Failed", str(error), parent=self.root)

    # ------------------------------------------------------------------ #
    # Persistence
    # ------------------------------------------------------------------ #
    def save_data(self):
        try:
            path = save_hotel(self.hotel)
            self.status_var.set(f"Saved to {path}")
            messagebox.showinfo("Saved", f"Hotel data saved to:\n{path}", parent=self.root)
        except Exception as error:
            messagebox.showerror("Save Failed", str(error), parent=self.root)

    def load_data(self, confirm=True):
        if confirm and not messagebox.askyesno(
                "Load Data",
                "Load saved data from disk?\nUnsaved changes will be replaced.",
                parent=self.root):
            return
        try:
            self.hotel = load_hotel(DATA_FILE)
            self.title_label.configure(text=self.hotel.name)
            self.root.title(f"{self.hotel.name} - Hotel Management System")
            self.refresh_all()
            self.status_var.set(f"Loaded from {DATA_FILE}")
        except Exception as error:
            messagebox.showerror("Load Failed", str(error), parent=self.root)

    # ------------------------------------------------------------------ #
    # Refresh everything
    # ------------------------------------------------------------------ #
    def refresh_all(self):
        self.refresh_dashboard()
        self.refresh_rooms()
        self.refresh_people()
        self.refresh_services()
        self.refresh_reservations()
````

#### Challenges Encountered


- Managing Room Availability
The system had to ensure that a reserved room could not be reserved again. This was solved by updating the room status immediately when a reservation object is created.

- Preventing Duplicate Reservations and state management.
A guest could accidentally make multiple active reservations for the same room. This was solved by checking existing active or pending reservations before creating a new one. Solved by triggering `room.make_reserved()` inside the `Reservation` constructor.

- Payment Tracking
The system needed to support partial payments and full payments. This was solved by adding payment tracking fields to the Reservation class and calculating balance due dynamically.

- Polymorphic Billing 
Getting different room types to calculate totals uniformly. Solved by enforcing an abstract `calculate_cost()` method in the base `Room` class.

- Object Serialization
 Converting nested OOP objects (Reservations containing Payments) into JSON was challenging. Solved by writing custom `to_dict()` and `from_dict()` methods.
